# Bruk av mengder (set)
Mengder er en itererbar datatype for å holde på en samling data
- lagrer unike elementer, ignorerer duplikater
- kan bare lagre ikke-muterbare datatyper (så vi kan ikke ha en mengde av lister, f.eks.)
- effektiv gjenfinning og bestemme om noe ikke finnes i mengden

Menger er nyttig for mange ting: 
- telle unike forekomster eller fjerne duplikater
- effektiv bestemme om en verdi er med eller ikke

Vi kan iterere over elementene i en mengde, men det er opp til Python-tolkeren i hvilken rekkefølge de kommer i.
Vil vi ha de f.eks. sortert, så kan vi lage en sortert liste fra mengden

**Noen Operasjoner på mengder** 
Merk at vi har muterende og ikke-muterende operasjoner som returnerer nytt objekt

For mer se https://eksamensvedlegg.it.ntnu.no/Felles/python.3107/library/stdtypes.html?highlight=set#set

| Muterende (endrer mengden) | Ikke-muterende (returnerer ny mengde) |
| --- | --- |
| `A.add(x)`    |   |
| `A.discard(x)` |  |
| `A.update(B)` | `A.union(B)` | 
| `A.intersection_update(B)` | `A.intersection(B)` | 
| `A.difference_update(B)` | `A.difference(B)` |
| `A.symmetric_difference_update(B)`  | `A.symmetric_difference(B)`  |

Noen viktige egenskaper:
- Alle elementene må ha samme type (heltall, flyttall, tekst, osv) 
- Elementene er muterbare, men ikke antall elementer
- Array kan reorganiseres med reshape (intern er en array bare en liste med fast lengde)
- Mer generell indeks-notasjon A[1,2] : element i rad 2, kolonne 3
- Arrays har en *shape*, som angir dimensjonene


Vi må bruke andre operasjoner for å bygge nye arrays, f.eks.
**konkatenering (sette sammen):**
- `np.hstack([A, B])` - setter sammen arrays horisontalt (side ved side)
- `np.vstack([A, B])` - setter sammen arrays vertikalt (over/under)
- `np.append(A, B)` - legger til B på slutten av A (flater ut til 1D)


```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
AcB = np.concatenate([A, B])  # → array([1, 2, 3, 4, 5, 6])
print(AcB)

C = np.array([[1, 2], [3, 4]])
D = np.array([[5, 6], [7, 8]])
CvD = np.vstack([C, D])  # → array([[1, 2], [3, 4], [5, 6], [7, 8]]):
print(CvD)
```

Numpy arrays støtter elementvise operasjoner (element-by-element):

Aritmetiske operasjoner:

A + B - addisjon elementvis
A - B - subtraksjon elementvis
A * B - multiplikasjon elementvis (IKKE matrisemultiplikasjon!)
A / B - divisjon elementvis
A ** 2 - potensering elementvis
A @ B - matrisemultiplikasjon
Sammenligningsoperasjoner:

A == B - returnerer boolean array med True/False for hvert element
A > B, A < B, A >= B, A <= B - sammenligner elementvis
np.all(A > 0) - True hvis alle elementer er > 0
np.any(A > 0) - True hvis minst ett element er > 0

**Numpy har noe som heter "broadcasting", som betyr at den kan i noen tilfeller tilpasse arrays av forksjellig form


```python
import numpy as np

A = np.array([1, 2, 3, 4])
B = np.array([10, 20, 30, 40])

A + B      # → array([11, 22, 33, 44])
A * 2      # → array([2, 4, 6, 8])
A ** 2     # → array([1, 4, 9, 16])
A > 2      # → array([False, False, True, True])
```

## Broadcasting
Numpy har broadcasting: Hvis man anvender (støttede) regneoperasjoner mellom arrays av ulik form,
vil numpy automatisk "strekke" den mindre arrayen for å matche den større.

Regler for broadcasting:

- Hvis arrays har ulikt antall dimensjoner, legg til dimensjoner (med størrelse 1) foran den minste
- Arrays er kompatible hvis dimensjonene enten er like, eller en av dem er 1
- Den minste arrayen "strekkes" langs dimensjoner med størrelse 1

```python
# Skalar med array - skalaren "strekkes" til alle elementer
A = np.array([1, 2, 3, 4])
A + 10  # → array([11, 12, 13, 14])

# 1D array med 2D array - 1D array gjentas for hver rad
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([10, 20, 30])
A + B  # → array([[11, 22, 33],
       #           [14, 25, 36]])

# Kolonne + rad = matrise
col = np.array([[1], [2], [3]])  # 3×1
row = np.array([10, 20, 30])      # 1×3
col + row  # → array([[11, 21, 31],
           #           [12, 22, 32],
           #           [13, 23, 33]])  # 3×3
```

## Akser (axes) i numpy arrays
Akser er dimensjonene i en array, nummerert fra 0:

axis=0: Første dimensjon (rader i 2D array, "nedover")
axis=1: Andre dimensjon (kolonner i 2D array, "bortover")

Huskeregel:
- axis=0 → "kollapser radene" (opererer på tvers av rader)
- axis=1 → "kollapser kolonnene" (opererer på tvers av kolonner)

```python
import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6]])

# A har shape (2, 3) - 2 rader, 3 kolonner

np.sum(A, axis=0)  # → array([5, 7, 9])    # Summerer nedover (langs radene)
np.sum(A, axis=1)  # → array([6, 15])       # Summerer bortover (langs kolonnene)
```

```python
print(A)
print(A.max(), A.sum())
```

```python
A = np.arange(24).reshape(2,3,4)
print(A.sum(axis=0))
print(A.sum(axis=1))
```

## Nøstede løkker
Når man har fler-dimensjonale lister eller arrays, vil det ofte dukke opp behov for å 
bruke nøstede løkker, f.eks.
- ytre løkke itererer over radene (første indeks)
- indre løkke itererer over kolonner, hvor langt ut i hver rad (andre indeks)

```python
# Eksempel på dobbel løkke med 2-dim array: Multiplikasjonstabell modulo 5

C = np.zeros((5, 5), dtype=int)
for i in range(5):
    for j in range(5):
        C[i, j] = (i * j) % 5
print(C)

# I følgende eksempel er indre løkke avhengig av den ytre løkken
for i in range(5):
    for j in range(i):
        C[i,j] *= 2
    for j in range(i, 5):
        C[i,j] = -C[i,j]
print(C)
```
