### Eksempel 1 - `for()` løkke

```python
# Dette er en "tuple". Vi skal se mer på det senere
sequence = (1, 3, 5, 7, 9, 11, 13)
for number in sequence:
    print(number)
```

### Eksempel 2 - `range()`

 - `range(stop)`
 - `range(start, stop)`
 - `range(start, stop, step)`

```python
for i in range(5):
    print(i)
```

```python
for j in range(5, 10):
    print(j)
```

```python
for k in range(0, 20, 2):
    print(k)
```

```python
for i in range(4):
    print("ITGK")
print("Ferdig!")
```

### Eksempel - `while()` løkke

Kjører kode **inntill en betingelse ikke lenger er oppfylt**

```python
i = 0
# Printer "Hello" 5 ganger
while i < 5:
    print("Hello")
    i += 1
```

```python
i = 5
# Printer tallene fra 5 til 10
while i <= 10:
    print(i)
    i += 1
```

```python
i = 0
while i < 4:
    print("ITGK")
    i += 1
print("Ferdig!")
```
