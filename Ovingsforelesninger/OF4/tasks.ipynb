### Eksempel - Plotting
Vi har en funksjon: $f(x)=x^3-2x^2+3$

**Hvordan plotte denne i python?**

````python
# 1. Importere bibliotek
import numpy as np
import matplotlib.pyplot as plt
````

````python
# 2. Definere funksjon 
def f(x):
  return x ** 3 - 2 * x ** 2 + 3
````

````python
# 3. Lage x-akse (bestem endepunkter og oppløsning)
x_verdier = np.linspace(-2, 2, 5)
print(x_verdier)
````

````python
# 4. Regne ut y-verdier (bruk funksjonen vi lagde)
y_verdier = f(x_verdier)
print(y_verdier)
````

````python
# 5. Plotte verdiene med Matplotlib
# plt.plot(x_verdier,y_verdier) # Viser verdiene som en sammenhengende linje
plt.plot(x_verdier,y_verdier, "o") # Viser verdiene som diskrete punkter
plt.show()
````
