### a) 
Lag en funksjon ```average_temperature(lst)``` som tar inn en 2d liste med temperatur for hver time over flere dager. Hvert element i den ytre listen representerer en dag, og hvert element i den indre listen representerer en time i løpet av dagen. 

Funksjonen skal returnere gjennomsnittstemperaturen for hver dag i form av en liste.

````python
temperature = [[6, 5, 10, 6, 5, 10, 10, 12, 15, 10, 14, 13, 15, 16, 17, 19, 18, 18, 13, 12, 9, 7, 6, 5], # Dag 1
               [4, 5, 5, 7, 9, 11, 12, 12, 13, 10, 10, 11, 12, 14, 17, 17, 15, 14, 13, 12, 11, 10, 9, 9], # Dag 2
               [8, 8, 9, 11, 10, 12, 15, 16, 17, 18, 19, 20, 21, 19, 20, 18, 17, 17, 15, 14, 11, 13, 12, 11]] # Dag 3

def average_temperature(lst):
  average = []
  for day in lst:
    sum = 0
    for hour in day:
      sum += hour
    average.append(sum/len(day))
  return average

print(average_temperature(temperature))
````

### b)
Lag en funksjon ```print_highest_temerature(lst)``` som printer når det var varmest gitt 2d-listen fra forrige oppgave.

```python
print_highest_temerature(temperature) # Det var varmest dag 3 kl. 12:00. Temperaturen var i det tidspunktet 21 grader. 
```

````python
import math

def print_highest_temperature(lst):
  highest = -math.inf
  day = 0
  hour = 0
  for day_index, day_value in enumerate(lst):
    for hour_index, hour_value in enumerate(day_value):
      if hour_value > highest:
        highest = hour_value
        day = day_index
        hour = hour_index

  print(f"Det var varmest dag {day + 1} kl. {hour}:00. Temperaturen var i det tidspunktet {highest} grader")

print_highest_temperature(temperature)
````

````python
def find_letter(word, letter):
    indices = []

    for i in range(len(word)):
        if word[i] == letter:
            indices.append(i)
    
    return indices

test_string = "Dette"
print(find_letter(test_string, "e"))
print(find_letter(test_string, "t"))
````

## Oppgave 4

Lag funksjonen ```is_angry(word)``` som tar inn en streng word og som returnerer ```True``` hvis alle alfabetiske tegn er store og strengen slutter på ‘!’.

Eksempel på kjøring:
```python
print(is_angry("Så fin du er!")) # Printer False
print(is_angry("SÅ FIN DU ER!")) # Printer True
```


````python
# Metode 1
def is_angry(streng):
  is_upper = streng == streng.upper()
  has_exclamation = streng[-1] == "!"
  return is_upper and has_exclamation

print(is_angry("Så fin du er!")) # Printer False
print(is_angry("SÅ FIN DU ER")) # Printer False
print(is_angry("SÅ FIN DU ER!")) # Printer True
````

````python
# Metode 2
def is_angry(word):
  if word[-1] != "!":
    return False

  for c in word:
    if c.isalpha() and c.islower():
      return False
    
  return True

print(is_angry("Så fin du er!")) # Printer False
print(is_angry("SÅ FIN DU ER")) # Printer False
print(is_angry("SÅ FIN DU ER!")) # Printer True
````

## Oppgave 5

Lag funksjonen `find(str, sub_str)` som tar inn to strenger og returnerer indexen til første forekomsten av `sub_str` i `str`. Dersom det er ingen forekomster, returner `-1`.

**Merk:** du skal ikke bruke innebygde funksjoner som finnes fra før av.

Eksempel på kjøring:
```python
print(find("Hei på deg!", "på")) # Printer 4
print(find("Hei på deg!", "yey")) # Printer -1
```

````python
def find(str, sub_str):
  lengde = len(sub_str)

  for i in range(len(str) - lengde + 1):
    if str[i:i + lengde] == sub_str:
      return i
  return -1

print(find("hei på deg!", "på"))
````

## Oppgave 6

Lag en funksjon `compress(text)` som tar inn en streng og erstatter påfølgende repeterende tegn med en enkelt forekomst av tegnet etterfulgt av antall repetisjoner.

Eksempel på kjøring:
```python
print(compress(“aaaaabcc”)) # Printer 5a1b2c
```

**Extra:** Hvis et tegn bare forekommer én gang, skal antallet ikke inkluderes.

````python
def compress(text):
  if text == "":
    return ""

  current_char = text[0]
  count = 1
  result = ""

  for i in range(1, len(text)):
    if text[i] == current_char:
      count += 1
    else:
      result += f"{count}{current_char}"
      count = 1
      current_char = text[i]

  # Legg til siste tegn
  result += f"{count}{current_char}"

  return result

print(compress("aaaaabcc"))
````

### Oppgave 2.1

Lag funksjonen ``home_draw_away`` som har en input-parameter ``matches`` som er en liste som
inneholder informasjon om fotballkamper i siste runde ble hjemmeseier ("H"), uavgjort ("U")
eller borteseier ("B").

Funksjonen skal returnere ei liste med 3 tall, der det første tallet er antallet hjemmeseire, det
andre tallet er antallet uavgjort, og det siste tallet er antallet borteseiere. Funksjonen skal
fungere på lister av ulik lengde, skal fungere for stor og liten bokstav for H, U og B, samt at
den skal ignorere resultat som ikke er H, U, eller B.


````python
matches = ("H", "u", "B", "h", "b", "H", "U", "H", "h", "B", "H", "U", "x", "h", "U")

def home_draw_away(matches):
    home, away, draw = 0, 0, 0

    for result in matches:
        result = result.upper()

        if result == "H":
            home += 1
        elif result == "U":
            draw += 1
        elif result == "B":
            away += 1

    return [home, draw, away]

print(home_draw_away(matches))
````

Lag funksjon ``pos_vowels`` som tar inn en parameter ``text`` som er en tekststreng og returnerer
posisjonen til alle norske vokaler i strengen ``text`` som en liste. Funksjonen skal fungere for
vokaler skrevet med stor og liten bokstav.

Eksempel på kjøring av kode:

```python
test = "DettE er et eksempel pÅ lAng liste med vokaler"
pos = pos_vowels(test)
print(pos)
# [1, 4, 6, 9, 12, 15, 18, 22, 25, 30, 33, 36, 40, 42, 44]
```

````python
test = "DettE er et eksempel pÅ lAng liste med vokaler"

vowels = "aeiouyøæå"

def pos_vowels(streng):
  pos = []
  for i, chr in enumerate(streng):
    if chr.lower() in vowels:
      pos.append(i)

  return pos

print(pos_vowels(test))
````

### Oppgave 2.3

Lag funksjon ``unique`` som har en inn-parameter ``text`` som er en tekststreng.

Funksjonen skal returnere en liste med alle unike ord i teksten skrevet med små bokstaver.
Rekkefølgen ordene i lista har er ikke bestemt. Merk at funksjonen skal returnere kun ord og
tegn som man typisk har i en setning skal fjernes. Man kan anta at tekstrengen starter med en
bokstav, at ordene er adskilt med mellomrom, at man ikke har tegn inne i et ord, og at man
ikke har flere enn et tegn etter hverandre.

Eksempel på kjøring av kode:
```python
test = "Er dette greit, eller er det IKKE greit? Det er greit tror jeg!"
words = unique(test)
print(words) 
# ["er", "dette", "greit", "eller", "det", "ikke", "tror", "jeg"]
```

````python
test = "Er dette greit, eller er det IKKE greit? Det er greit tror jeg!"
tegn = '.,:;?!-_*@+'

def unique(streng):
  words = streng.lower().split()
  unique = []
  for word in words:
    word = word.strip(tegn)
    if word not in unique:
      unique.append(word)
  return unique

print(unique(test))
````

Oppgave 7 – Cæsarchiffer

Innen kryptografi er cæsarchiffer en av de enkleste og mest kjente krypteringsteknikkene.
Hver bokstav i klarteksten erstattes med en annen bokstav et gitt antall steg lenger frem
i alfabetet. Med et skift på 1 blir A til B, B til C, og så videre.

Lag funksjonen krypter(tekst, skift) som krypterer en tekst med cæsarchiffer.
Parameteren skift forteller hvor mange steg hver bokstav forflytter seg i alfabetet.

Funksjonen er noe forenklet:
- Den skal kun ta hensyn til bokstavene i det norske alfabetet:
  ABCDEFGHIJKLMNOPQRSTUVWXYZÆØÅ
- Etter Å fortsetter alfabetet på A igjen.
- Funksjonen skal returnere den krypterte teksten med store bokstaver.
- Alle mellomrom skal være fjernet.
- Du kan anta at teksten kun inneholder bokstaver fra det norske alfabetet og mellomrom.

Eksempler på kjøring:
print(krypter("Kryptering er gøy", 1))  # Printer LSZQUFSJOHFSHÅZ
print(krypter("Zebra", 1))              # Printer ÆFCSB
print(krypter("Å", 1))                  # Printer A

Hint:
- Lag en streng med alfabetet og bruk .index() eller find() for å finne posisjonen til en bokstav.
- Bruk .upper() og .replace(" ", "") (eller hopp over mellomrom i løkka).
- Bruk modulo (%) med lengden på alfabetet for å "gå rundt" etter Å.

````python
def krypter(tekst, skift):
    alfabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZÆØÅ"
    tekst = tekst.upper()
    resultat = ""

    for tegn in tekst:
        if tegn != " ":
            indeks = alfabet.index(tegn)
            ny_indeks = (indeks + skift) % len(alfabet)
            resultat += alfabet[ny_indeks]

    return resultat


print(krypter("Kryptering er gøy", 1))  # Skal printe LSZQUFSJOHFSHÅZ
print(krypter("Zebra", 1))              # Skal printe ÆFCSB
print(krypter("Å", 1))                  # Skal printe A
````
