## 5. Andre måter å lagre flere registreringer

I stedet for å ha en (lang) liste med ett element for hver registrering,
så kan det være mer økonomisk å *telle* antall registreringer per person.

Oppgave: Legg til kommentarer i koden under som forklarer hva den gjør

````python
påmeldinger = ["Ada", "Bob", "Cher", "Bob", "Ada"]  # 

deltagere = list(set(påmeldinger)) # 
antall = [påmeldinger.count(deltager) for deltager in deltagere] #

if "David" in deltagere:
    index = deltagere.index("David")
    antall[index] += 1
else:
    deltagere.append("David")
    antall.append(1)
````

### 6. Noen ulemper med koden over

- Vi må holde orden på to lister, og passe på å synkronisere disse
- Det er (relativt) lite effektivt å lete opp navn i en lang liste, og 
- vi kan ikke bruke en mengder for navn hvis vi skal bruke indeks for oppslag i den andre listen


For å slippe å ha to separate lister, så kunne vi også brukt en liste med navn og antall sammen, f.eks. slik:
```python
påmeldinger = [["Ada", 2], ["Bob", 2], ["Cher", 1]]
```

Her er det imidlertid litt knotete å søke opp navn i en slik liste. Vi skal ikke *knote* med det her, i stedet skal vi se
at Python har en perfekt løsning som kombinerer navnene som i en mengde, og med tilhørende data (antall)

## 7. Ordbøker (dictionary): Set med opplsagsnøkler + data

Vi kan tenke på en dictionary som en mengde (set) av oppslagsnøker, 
og hvor hvert element har en tihørende verdi/data-objekt.

Vi bruker *nøklene* i stedet for indeks for å få dataene.

I eksemplet vi har jobbet med, skal vi bruke navn som *nøkkel*,
og antallet påmeldinger som tilhørende verdi

Se på koden under. Kan du gjette hva som skjer i hver linje?

````python
reg_dict = {"Ada":2, "Bob":2, "Cher":1}

print(reg_dict.get("Ada"))  #
luke_reg = reg_dict.get("Luke", 0)  #  
print(luke_reg)

# Hint: Navnet på metoden setdefault kan virke litt misvisende
# Hva er forskjellen på get og setdefault? Skriv svaret i kommentaren under.
dora_reg = reg_dict.setdefault("Dora", 0)  #
ada_reg = reg_dict.setdefault("Ada", 0)  #

print(reg_dict)
````

Vi ser at vi både dict.get() og dict.setdefault() gir oss tilgang til verdien tilhørende en nøkkel, 
men kun `setdefault()` oppretter et element i ordboken hvis den ikke finnes.

Vi har ikke vanlige indekser 0, 1, 2, ... som for lister, men vi bruker nøklene i stedet:

````python
if "David" in reg_dict:  # Sjekker på nøkler, ikke verdier.
    print(f"David er {reg_dict['David']} ganger fra før") 
    reg_dict["David"] += 1
else:
    print("David er ikke registrert fra før, legger til med 1")
    reg_dict["David"] = 1  

# Ved tilordning, så vil nøkkel opprettes hvis den ikke finnes fra før. Derfor kan vi gjøre følgende uten å sjekke først:
reg_dict["Bob"] = reg_dict.get("Bob", 0) + 1 

print(reg_dict)
````

## 8. Muterbare og ikke-muterbare verdier i ordbøker

**NB! Elementene i mengder og nøkler i ordbøker må være ikke-muterbare.**

*Verdiene* i en orbok kan være muterbare. En verdi som er ikke-muterbar, kan kun endres ved *ny tilordning* som for alle
andre ikke-muterbare datatyper:

```python
ordbok[nøkkel] = ny_verdi
```

Vi har også en metode som kan sette eller oppdatere flere elementer på en gang fra en annen ordbok:

```python
ordbok.update({key1: value1, key2: value2})
``` 

Hvis verdien er *muterbar*, så kan vi bruke referansen vi får fra get(), setdefault() for å gjøre oppdateringer:


````python

venner = {"Ada":["Bob", "Cher"], "Bob":["Ada"], "Cher":["Ada"]}
venner["David"] = ["Cher"]
venner["Cher"].append("David")  

venner.setdefault("Xander", []).append("Ada")  
venner.setdefault("Ada", []).append("Xander")
print(venner)

# Men følgende vil ikke fungere, 
tull = {1: 1, 2: 2}
tull.setdefault(3,0) + 1
print(tull)
````
