### Flere parametre som tuppel

FACT: Mange funksjoner har flere parametre, dvs. tar flere argumenter. Noen tar til og med et *variabelt* antall argumenter,
som f.eks. print. Hvis vi sender en liste eller tuppel til print, så vil det dog tolkes som kun ett argument.

Hvis vi vil at en liste eller tuppel skal tolkes som flere argumenter, så kan vi *pakke ut* en med *-operatoren

````python
tuppel = ("Ada", "Bob", "Cher", "David")

print(tuppel) # ett argument, som viser hele tuppelen som ett objekt
print(*tuppel) # Skriv ut alle elementene i tuppelen, separert med mellomrom
print("Ada", "Bob", "Cher", "David") # gir akkurat det samme som forrige linje
````

Hvis vi ser på dokumentasjonen til print, så ser vi hvordan vi definerer en funksjon slik
at den kan ta inn et variabel antall argumenter: 

```python
def print(
    *values: object,
    sep: str | None = " ",
    end: str | None = "\n",
    file: SupportsWrite[str] | None = None,
    flush: Literal[False] = False
) -> None
```

Under har vi laget en enkel funksjon som bruker det samme:

````python
def min_print(*args):
    # alle argumentene som sendes inn blir samlet i en tuppel som heter args
    print(type(args))
    for arg in args:
        print(arg, end=" - ")
    print()  # for å legge til linjeskift etter utskriften

min_print("Ada", "Bob", "Cher", "David") # Skriv ut alle elementene i tuppelen, separert med mellomrom
````

## zippe lister

zip() kombinerer flere itererbare objekter til én iterator av tupler:

````python
navneliste = ["Ada", "Bob", "Cher", "David"]

zip_objekt = zip(navneliste, range(1, len(navneliste)+1)) # lager en zip-objekt som inneholder tupler med navn og tall
print(zip_objekt) # viser at det er et zip-objekt, ikke en liste eller tuppel

for element in zip_objekt:
    print(element) # viser hvert element i zip-objektet, som er en 2-tuppel bestående av navn og tall  

# Vi kan zippe så mange objekter vi vil, men de vil bare zippe så langt som det korteste objektet. 
mange_zip = zip(navneliste, range(1, len(navneliste)+1), "noe tekst her")

for element in mange_zip:
    print(element)
````

````python
# Triksing med utpakking av lister og zip. Kan du se hva som skjer her?

tall_bokstaver = list(zip(range(1,6), "abcde"))  # [(1, 'a'), (2, 'b'), (3, 'c'), (4, 'd'), (5, 'e')]

tall2, bokstaver2 = zip(*tall_bokstaver)  # tall2 = (1, 2, 3, 4, 5), bokstaver2 = ('a', 'b', 'c', 'd', 'e') 

print(tall2, bokstaver2)  # (1, 2, 3, 4, 5) ('a', 'b', 'c', 'd', 'e')
````

## Mer om `zip` og utpakking: tenk først, kjør etterpå

**Arbeidsmåte:** Skriv ned hva du tror som svar på spørsmålene i hver celle *før* du kjører koden. 

Poenget er å bygge en mental modell, ikke å huske syntaks. Dette er "snacks": ikke eksamensstoff, men nyttig å kjenne igjen.

### 1. Hvorfor `zip`? Parallelle lister uten indekser

Vi har to lister der plassene hører sammen.

````python
navn  = ["Ada", "Bob", "Cher"]
poeng = [2, 5, 1]

# Med indekser:
for i in range(len(navn)):
    print(f"{navn[i]}: {poeng[i]}")

# Med zip. Hva tror du er forskjellen i lesbarhet?
for n, p in zip(navn, poeng):
    print(f"{n}: {p}")
````

**Spørsmål (gjør dette før du kjører neste celle):**

1. Hva gir `list(zip(navn, poeng))`?
2. Hva skjer hvis `poeng` bare har to elementer?
3. Hva gir `dict(zip(navn, poeng))`? Hvilken sammenheng har dette med ordbøker?

*Metafor:* en glidelås (engelsk *zipper*) fletter to rader med tenner sammen, ett par om gangen.

````python
print(list(zip(navn, poeng)))
print(list(zip(navn, poeng[:2])))      # zip stopper ved den korteste
print(dict(zip(navn, poeng)))
````

### 2. Utpakking: tilordne flere variabler på én gang

Når høyresiden er en tuppel eller liste, kan vi gi hvert element sitt eget navn.

**Spørsmål:** Hva er forskjellen mellom `for par in zip(navn, poeng)` og `for n, p in zip(navn, poeng)`?

````python
a, b = (1, 2)
print(a, b)

for par in zip(navn, poeng):
    print(par, type(par))     # én variabel: hele tuppelen

for n, p in zip(navn, poeng):
    print(n, p)               # utpakket: to variabler
````

### 3. `*` på venstre side: "ta resten"

**Spørsmål:** Hva blir `resten` hvis listen bare har ett element? Hva blir den hvis `*` står først?

````python
første, *resten = [1, 2, 3, 4]
print(første, resten)

*alle_unntatt_siste, siste = [1, 2, 3, 4]
print(alle_unntatt_siste, siste)

x, *resten = [1]
print(x, resten)               # resten er en tom liste
````

### 4. `*` ved funksjonskall: pakk ut som separate argumenter

Dette er det motsatte av `*args` i definisjonen. Med `*args` samler vi argumenter i en tuppel. Med `*` i kallet pakker vi ut en samling til argumenter. `**` gjør det samme for ordbøker og navngitte argumenter.

**Spørsmål:** Hva tror du `legg_sammen(*tall)` er det samme som?

````python
def legg_sammen(a, b, c):
    return a + b + c

tall = [1, 2, 3]
print(legg_sammen(*tall))

person = {"a": 10, "b": 20, "c": 30}
print(legg_sammen(**person))
````

### 5. Det "magiske" eksempelet: `zip(*par)` snur en tabell

Tenk deg listen av par som en tabell med to kolonner:

| navn | poeng |
|---|---|
| Ada  | 2 |
| Bob  | 5 |
| Cher | 1 |

`zip(*par)` er det samme som `zip(("Ada", 2), ("Bob", 5), ("Cher", 1))`. `zip` plukker første element fra hver tuppel, så andre element fra hver tuppel. Rader blir til kolonner.

**Spørsmål:**

1. Skriv ned hva `navn_t` og `poeng_t` blir før du kjører.
2. Hva tror du `list(zip(*matrise))` gir for en matrise med 2 rader og 3 kolonner?

````python
par = [("Ada", 2), ("Bob", 5), ("Cher", 1)]
navn_t, poeng_t = zip(*par)
print(navn_t)
print(poeng_t)

matrise = [[1, 2, 3], [4, 5, 6]]
print(list(zip(*matrise)))     # transponert matrise
````

### 6. Vanlige fallgruver

**Spørsmål:** Hvorfor skriver den siste linjen ut en tom liste?

````python
z = zip(navn, poeng)
print(z)               # viser bare et zip-objekt, ikke innholdet
print(list(z))         # nå brukes iteratoren opp
print(list(z))         # tom: en zip kan bare itereres én gang
````

**Spørsmål:** Hva skriver koden under ut? En `zip` tar ikke kopi av listene, bare iteratorer over dem. Hva skjer hvis vi endrer listene før zip-en er ferdig?

````python
a = [1, 2, 3, 4]
b = ["a", "b", "c", "d"]
z = zip(a, b)
print(next(z))
a[1] = 99              # endrer et element som ikke er hentet ennå
print(list(z))
````

````python
a = [1, 2, 3, 4]
b = ["a", "b", "c", "d"]
z = zip(a, b)
print(next(z))
a.pop(0)               # fjerner foran gjeldende posisjon: alt forskyves
print(list(z))
````

**Forklaring:** Listeiteratoren husker bare en indeks og slår opp i den levende listen. Endrede elementer som ikke er hentet ennå, vises. Fjerning eller innsetting foran posisjonen forskyver og hopper over elementer. Utvider du den korteste listen, fortsetter `zip`. Ingen feilmelding, i motsetning til når man endrer størrelsen på en ordbok under iterasjon.

**Råd:** Ikke endre listene mens du bruker `zip`. Trenger du et øyeblikksbilde, bruk `list(zip(a, b))`.

**Oppsummering**

- `zip(a, b)` setter sammen element for element og stopper ved den korteste samlingen.
- `a, b = par` pakker ut en tuppel eller liste til variabler. `*rest` tar "resten".
- `f(*samling)` pakker ut en samling til argumenter. `f(**ordbok)` pakker ut en ordbok til navngitte argumenter.
- `zip(*rader)` transponerer: rader blir kolonner.
- `zip` gir en iterator som bare kan brukes én gang. Bruk `list(...)` for å se innholdet.
