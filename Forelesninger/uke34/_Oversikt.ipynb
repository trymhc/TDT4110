# Oversikt uke 34

Her finner du en kort innholdsfortegnelse over notatbokene for uke 34. Bruk denne som en liten "startpakke" for a bygge flyt i programmering fra forste linje kode.

## Innholdsfortegnelse

1. [BP_00_input_print.ipynb](BP_00_input_print.ipynb)  
   Grunnleggende bruk av `input()` og `print()`: hvordan programmet kan ta imot data og gi tydelige svar tilbake.

2. [BP_01_plotting.ipynb](BP_01_plotting.ipynb)  
   Introduksjon til plotting: visualiser data og se hvordan tall blir til figurer som er lettere a forsta.

3. [FP_00_Aritmetiske_uttrykk.ipynb](FP_00_Aritmetiske_uttrykk.ipynb)  
   Aritmetiske uttrykk i Python: regneoperasjoner, rekkefolge og hvordan uttrykk evalueres.

4. [FP_01_tilordning_intro.ipynb](FP_01_tilordning_intro.ipynb)  
   Tilordning i praksis: hva variabler er, og hvordan verdier lagres og oppdateres i kode.

5. [FP_02_Tilordning_objektreferanser.ipynb](FP_02_Tilordning_objektreferanser.ipynb)  
   Objektreferanser forklart enkelt: forskjellen pa kopier og referanser, og hvorfor det betyr noe.

6. [FP_03_intro_funk.ipynb](FP_03_intro_funk.ipynb)  
   En myk start pa funksjoner: hvorfor vi bruker dem, og hvordan de gir mer ryddig og gjenbrukbar kode.

7. [FP_04_SyntaxError.ipynb](FP_04_SyntaxError.ipynb)  
   Vanlige syntaksfeil og hvordan du leser feilmeldinger for a rette dem raskt.

8. [FP_05_NameError.ipynb](FP_05_NameError.ipynb)  
   Fokus pa `NameError`: hva feilen betyr, nar den oppstar, og gode strategier for feilsoking.

9. [PED_00_kodetraversering.ipynb](PED_00_kodetraversering.ipynb)  
   Kodetraversering steg for steg: tren pa a "kjoere" kode i hodet for a forsta flyten bedre.

## Tips til rekkefolge

For best progresjon kan du starte med FP-notatbokene, ta BP-notatbokene underveis for praktisk trening, og avslutte med PED for dypere kodeforstaelse. Litt og ofte gir stor framgang!
