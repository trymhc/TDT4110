# Kodetraversering: Hva gjør Python?

I denne notebooken trener vi på å følge et program linje for linje.

For hvert eksempel skal du:

1. lese koden ovenfra og ned
2. skrive ned verdiene til variablene
3. forutsi hva som blir skrevet ut
4. kjøre koden og kontrollere svaret

Målet er ikke bare å vite *hva* svaret blir, men å kunne forklare *hvordan* Python kom fram til svaret.

Vi skal dekke alle disse konseptene grundig etter hvert. Hvis du ikke har hatt noe Python fra før, så ta det som en smakebit. Har du hatt noe Python før, kan det fungere som litt oppfrisking. Det viktige er at vi forstår den grunnleggende virkemøten til Python, OG AT DET Å MANUELT GÅ GJENNOM KODE ER VIKTIG FOR Å FORSTÅ OG KUNNE SKRIVE EGEN KODE.

## 1. Python gjør én linje om gangen

Start med å lese koden uten å kjøre den.

Spør deg selv:

- Hva skjer på den første linjen?
- Hva skjer på den andre linjen?
- Hva tror du kommer i utskriften?

`print()` betyr at Python skal vise en verdi eller en tekst.

````python
print("Hei")
print("Velkommen til Python")
````

## 2. Variabler er navn på verdier

En tilordning lagrer en verdi under et navn. Python evaluerer høyresiden først, og lagrer deretter resultatet under navnet på venstresiden.

````python
navn = "Ada"
alder = 20

print(navn)
print(alder)
````

## Spor verdiene

Etter de to første linjene ser Python omtrent slik ut:

| Navn | Verdi |
|---|---|
| `navn` | `"Ada"` |
| `alder` | `20` |

Prøv å forutsi hva som skjer hvis vi legger til:

```python
navn = "Bo"
print(navn)
```

Den nye tilordningen endrer verdien som ligger under navnet `navn`.

## 3. Python beregner uttrykk

Når Python møter `c = a + b`, må uttrykket på høyre side beregnes før resultatet lagres i `c`.

````python
a = 3
b = 4
c = a + b

print(c)
````

## 4. En variabel kan få ny verdi

Spor dette nøye. Verdien til `alder` endres etter at den først ble lagret.

````python
alder = 20
alder = alder + 1

print(alder)
````

## Hva betyr `alder = alder + 1`?

Python gjør dette i to trinn:

1. Finn den gamle verdien til `alder` og legg til `1`.
2. Lagre den nye verdien under navnet `alder`.

Dette er en oppdatering av en variabel, ikke en matematisk likning.

## 5. Funksjoner

En funksjon samler instruksjoner under et navn. Når funksjonen kalles, hopper Python inn i funksjonen, gjør arbeidet, og kommer tilbake med en verdi.

````python
def doble(tall):
    resultat = 2 * tall
    return resultat

x = doble(4)
print(x)
````

## Spor funksjonskallet

Når Python utfører `x = doble(4)`:

1. `tall` får verdien `4`.
2. `resultat` blir `8`.
3. `return resultat` sender `8` tilbake.
4. `x` får verdien `8`.

`print()` viser en verdi. `return` sender en verdi tilbake fra en funksjon.

## 6. `if` velger mellom to muligheter

Python evaluerer betingelsen. Bare blokken som passer, blir kjørt.

Legg merke til innrykket.

````python
temperatur = 18

if temperatur < 0:
    beskjed = "Det er frost."
else:
    beskjed = "Det er ikke frost."

print(beskjed)
````

## Spor `if`-setningen

Her er `temperatur < 0` lik `False`, fordi `18` ikke er mindre enn `0`. Derfor kjører Python `else`-blokken.

Prøv å endre temperaturen til `-3` og forutsi resultatet før du kjører cellen på nytt.

## 7. En løkke gjentar en blokk

Nå kombinerer vi variabel, liste og gjentakelse. Spor verdien av `summe` for hver runde.

````python
summe = 0

for tall in [2, 5, 1]:
    summe = summe + tall

print(summe)
````

## Spor løkken

| Runde | `tall` | `summe` etter oppdatering |
|---|---:|---:|
| Før løkken |  | 0 |
| 1 | 2 | 2 |
| 2 | 5 | 7 |
| 3 | 1 | 8 |

Løkken gjør den innrykkede linjen én gang for hvert element i listen.
