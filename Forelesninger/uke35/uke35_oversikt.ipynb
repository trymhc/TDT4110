## Funksjoner - ikke bare matte

Funksjoner i Python har minst to viktige funksjoner:

- Å strukturere koden i deler slik at vi kan forholde oss til en del om gangen.
- Å kunne gjenbruke samme kode flere steder. Å måtte skrive samme kode om igjen tar tid, gjør koden mer uoversiktlig, og det er lettere å gjøre feil.

F.eks. når vi bruker funksjonene input() og print(), så trenger vi ikke tenke på hvordan de gjør ting, bare hva de gjør.det er samme kode som kjøres hver gang. Ved å bruke input til funksjonene, det vi kaller funksjonsargumenter, så blir funksjonene mer allsidige: Vi kan f.eks. velge hva som skal skrivet ut, og en del om hvordan det skal se ut.

Vi skal se hvordan en lager egne funksjoner og hvordan disse kan brukes. Vi skal se på hvordan funksjoner

- strukturerer koden vår
- hvordan programflyten blir
- hva funksjonen "vet om verden rundt seg" og hvordan den får og returnerer data.
