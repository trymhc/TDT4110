## `break` og `continue`

For hver kodebit under: Forutsi hva som skrives ut før du kjører cellen. Forklar deretter hva som skjer når programmet møter `break` eller `continue`.

```python
# Oppgave: Hva skrives ut? Når avsluttes løkka?
n = 0
while n < 6:
    n = n + 1
    if n == 4:
        break
    print(n)

print('etter løkka:', n)
```

```python
# Oppgave: Hva skrives ut? Hvilken verdi blir ikke skrevet ut?
n = 0
while n < 6:
    n = n + 1
    if n == 4:
        continue
    print(n)

print('etter løkka:', n)
```

## `for`-løkker

For hver kodebit under: Forutsi hva som skrives ut før du kjører cellen. Følg hvilken verdi løkkevariabelen får i hver runde, og legg spesielt merke til hva som er med og ikke med i `range`.

```python
# Oppgave: Hva skrives ut, og hvor mange ganger kjører løkka?
for tall in [3, 1, 4]:
    print(tall)
print('ferdig')
```

```python
# Oppgave: Hvilke verdier får n, og hvorfor skrives ikke 5 ut?
for n in range(1, 5):
    print(n)
print(n)
```

```python
# Oppgave: Hva er verdien av sum_ etter hver runde, og til slutt?
sum_ = 0
for tall in range(2, 5):
    sum_ = sum_ + tall
    print(sum_)
print('slutt:', sum_)
```
