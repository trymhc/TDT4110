## `while`-løkker

En `while`-løkke kjører kodeblokken på innrykk så lenge betingelsen er `True`. Betingelsen testes før hver runde, så løkka kan kjøre null ganger.

Se [while-løkker](FP_30_while.ipynb) for virkemåte, innrykk, telleroppdatering, brukerinput og `break`. Pass særlig på at minst én verdi som inngår i betingelsen blir endret, ellers kan du få en evig løkke.

```python
tall = 1
while tall <= 3:
    print(tall)
    tall += 1
print('ferdig')
```

I eksempelet over er `tall` både initialisert før løkka og oppdatert i hver runde. Følg rekkefølgen nøye: test betingelsen, kjør innrykket kode, oppdater `tall`, og test på nytt.

## `for`-løkker og sekvenser

En `for`-løkke går gjennom elementene i en sekvens, for eksempel en liste, en streng eller et array. Løkkevariabelen får automatisk neste verdi i hver runde.

Se [for-løkker](FP_31_for.ipynb) for sekvenser, tomme sekvenser, strenger, `range`, `numpy.arange`, `linspace`, `zeros` og `ones`. Innrykket viser hvilke linjer som hører til løkka.

```python
for navn in ['Ada', 'Grace', 'Katherine']:
    print('Hei', navn)
```

## `while` eller `for`?

- Bruk `while` når antall repetisjoner avhenger av en betingelse eller ikke er kjent på forhånd.
- Bruk `for` når du skal gå gjennom en kjent sekvens eller et bestemt intervall.

I [sammenligningen av `while` og `for`](FP_32_While_og_for-løkker__eksakt_forskjell_i_virkemåte-2.ipynb) kan du se forskjellen mellom automatisk iterasjon i `for` og eksplisitt oppdatering i `while`. Legg merke til hva som styrer neste runde, og hvordan variabler opprettes, oppdateres eller muteres.

## Løkker på tallserier

Løkker kan brukes til å bygge opp en sum eller beregne verdier i en tallserie. I [løkker og tallserier](BP_33_løkker_tallserier.ipynb) sammenlignes løkker med `range`, NumPy og Gauss-formelen. Dette gir også trening i å se når en løkke kan erstattes av en direkte eller vektorisert beregning.

```python
n = 5
sum_ = 0
for tall in range(1, n + 1):
    sum_ += tall
print(sum_)
```
