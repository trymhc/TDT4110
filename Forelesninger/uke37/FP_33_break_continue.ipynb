# Endre kodeflyten i løkker - break og continue

NB! Ikke les før du føler du forstår programflyten til løkker

I noen situasjoner er det nyttig å kunne enten

- hoppe ut av en løkke. Da bruker vi `break`
- hoppe videre til neste iterasjon fra inne i kode-blokken, før vi har kommet til slutten av den. Da bruker vi `continue`

## `break` - bryt ut av løkka

`break` avslutter løkka med en gang. Programmet fortsetter deretter på første linje etter løkka. Hvis en har nøstede løkker, så er det kun løkke som break-kommandoen står i som avbrytes. 

Dette er særlig nyttig når løkka har oppnådd målet sitt før den ellers ville ha stoppet, for eksempel når vi har funnet det første elementet vi leter etter.

Før du kjører eksempelet: Hvilke navn blir skrevet ut, og hva er verdien av `indeks` etter løkka?



```python
navn = ['Ada', 'Bjarne', 'Chen', 'Dina']
indeks = 0

while indeks < len(navn):
    if navn[indeks] == 'Chen':
        break
    print(navn[indeks])
    indeks += 1

print('Fant Chen på indeks', indeks)
```

Uten `break` måtte vi enten latt løkka gå gjennom resten av lista uten grunn, eller brukt en ekstra variabel i betingelsen.

## `continue` - hopp til neste iterasjon

`continue` hopper over resten av kodeblokken i den pågående runden og går direkte til neste "runde"
og fortsetter på vanlig måte

- For en while løkke, blir løkkeb-etingelsen sjekket før den fortsetter
- For en for-løkke, så hentes neste verdi hvis det er noen igjen

Dette er nyttig når enkelte verdier skal ignoreres, men resten skal behandles på samme måte.

Før du kjører eksempelet: Hvilket tall blir ikke skrevet ut? Hvorfor stopper ikke løkka når `n` er 4?



```python
# Koden under skriver ut tallene fra 1 til 5, men hopper over tallet 4.
# Dette gjøres ved å bruke continue, som gjør at koden hopper over resten av koden i løkken og går til neste iterasjon.

n = 0

while n < 6:
    n += 1
    if n == 4:
        continue
    print(n)

print('ferdig')
```

Uten `continue` kunne vi skrevet resten av løkkekroppen inni en `if n != 4`-blokk. `continue` gjør det tydelig at 4 er et unntak, mens resten av behandlingen er den vanlige.

```python
# Løkke som printer ut tallene fra 1 til 5, men hopper over tallet 4.

n = 0
while n < 6:
    n += 1
    if n != 4:
        print(n)
    
```

Oppgave: Hva skrives ut av koden under?

```python
n = 0
while n < 6 and n != 4:
    n += 1
    print(n)
```



I en `while`-løkke må du være ekstra oppmerksom på oppdateringen av løkkevariabelen. Hvis `continue` skjer før verdien endres, kan løkka bli evig. 

```python
# Eksempel på løkke som bruker continue, men som da ikke får oppdatert variabelen som brukes i while-betingelsen.
# Dette vil føre til en uendelig løkke. 

n = 1
while n < 6:
    if n == 4:
        continue
    print(n)
    n += 1
```
