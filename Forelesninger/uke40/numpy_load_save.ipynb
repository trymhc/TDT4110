## `savetxt` og `loadtxt`

De viktigste parameterne er:

### `np.savetxt(fname, X, ...)`

- `fname`: filnavn eller filsti
- `X`: arrayen som skal lagres
- `fmt`: format for tallene, for eksempel `"%.2f"` eller `"%d"`
- `delimiter`: skilletegn mellom kolonnene, for eksempel `","` eller `";"`
- `header`: tekst øverst i filen
- `footer`: tekst nederst i filen
- `comments`: tegn som settes foran `header` og `footer`, vanligvis `"# "`

### `np.loadtxt(fname, ...)`

- `fname`: filnavn eller filsti
- `dtype`: datatypen som skal leses, for eksempel `float` eller `int`
- `delimiter`: samme skilletegn som ble brukt ved lagring
- `skiprows`: hvor mange linjer som skal hoppes over fra toppen
- `usecols`: hvilke kolonner som skal leses
- `unpack`: med `True` får vi én array per kolonne
- `ndmin`: minste antall dimensjoner på resultatet

I eksemplene under lagres filene i notebookens arbeidsmappe, altså `Path.cwd()`.

```python
# En todimensjonal array med temperaturer for fire dager
maalinger = np.array([
    [1, 12.5],
    [2, 14.0],
    [3, 9.5],
    [4, 16.0],
])
```

```python
# Lagre med overskrift, komma mellom kolonnene og én desimal
np.savetxt(
    "temperaturer.csv",
    maalinger,
    fmt=["%d", "%.1f"],
    delimiter=",",
    header="dag,temperatur",
    comments="",
)
```

Sjekk at du får en fil med navn "temperaturer.csv". Åpne den og sjekk innholdet.

En .csv-fil er en vanlig tekstfil, men med innhold strukturert som en tabell,
hvor en bruker linjeskift for skille rader, og komma eller et annet tegn til å skille
elementene i hver rad.

```python
# Les inn igjen. Den første raden er overskriften, derfor skiprows=1.
innlest = np.loadtxt(
    "temperaturer.csv",
    delimiter=",",
    skiprows=1,
)

print("Innlest:")
print(innlest)
```

```python
# Lagre heltall med en enkel tekstoverskrift og avsluttende kommentar.

tall = np.array([[10, 20, 30], [40, 50, 60]])
np.savetxt(
    "tall.txt",
    tall,
    fmt="%d",
    delimiter=";",
    header="a;b;c",
    footer="slutt",
)

# Les bare kolonne 0 og 2 som heltall.
valgte_kolonner = np.loadtxt(
    "tall.txt",
    dtype=int,
    delimiter=";",
    skiprows=1,
    usecols=(0, 2),
)
print("Valgte kolonner:")
print(valgte_kolonner)

# unpack=True gir én array per kolonne.
a, b, c = np.loadtxt(
    "tall.txt",
    dtype=int,
    delimiter=";",
    skiprows=1,
    unpack=True,
)
print("Kolonnene:", a, b, c)

# ndmin=2 sørger for at resultatet minst blir en todimensjonal array.
forste_kolonne = np.loadtxt(
    "tall.txt",
    dtype=int,
    delimiter=";",
    skiprows=1,
    usecols=(0,),
    ndmin=2,
)
print("Form med ndmin=2:", forste_kolonne.shape)
```
