# Nøstede if-setninger

En `if`-setning kan inneholde en ny `if`-setning. Dette kalles **nøsting**. Den indre betingelsen blir bare undersøkt når den ytre betingelsen er sann.

I eksempelet nedenfor undersøker vi først om alderen er gyldig. Bare da er det nyttig å spørre om personen er gammel nok til å stemme.

````python
alder = 16

if alder > 0:
    if alder < 18:
        print("Kom tilbake til neste valg")
    else:
        print("Fortsett til valglokalet")
else:
    print("Ugyldig inngangsdata")
````

## Les innrykket

```python
if alder > 0:                 # ytre betingelse
    if alder < 18:            # indre betingelse
        print("For ung")
    else:
        print("Gammel nok")
else:
    print("Ugyldig alder")
```

- Alt som er rykket inn under den ytre `if`-setningen, hører til den.
- Når innrykket avsluttes, er også den ytre blokken slutt.
- `else`-setningen på samme innrykksnivå som den ytre `if`-setningen hører til den ytre testen.

Flere nivåer er mulig, men mye nøsting kan gjøre kode vanskelig å lese. Da kan `elif` eller boolske operatorer som `and` og `or` ofte være et bedre valg.

````python
alder = 20
er_beruset = True

if alder >= 18:
    if er_beruset:
        print("Du slipper ikke inn")
    else:
        print("Velkommen inn")
else:
    print("Du må være minst 18 år")
````
