# Uke 36: Betingelser og logiske uttrykk

Denne uken ser vi på hvordan programmer kan velge hva de skal gjøre ut fra en betingelse. Boolske uttrykk gir verdien `True` eller `False`, og `if`-setninger bruker slike uttrykk til å styre programflyten.

Målet er å kunne lese, skrive og forutsi virkningen av enkle betingelser i Python.

## Boolske uttrykk

Et boolsk uttrykk har verdien `True` eller `False`. Slike uttrykk lager vi blant annet med:

- sammenligningsoperatorer: `==`, `!=`, `<`, `>`, `<=` og `>=`
- medlemskapsoperatorer: `in` og `not in`
- identitetsoperatoren `is`, når vi vil undersøke om to navn viser til samme objekt
- boolske operatorer: `and`, `or` og `not`

Se [boolske uttrykk](FP_20_boolske-uttrykk.ipynb) for forklaringer, sannhetstabeller og oppgaver.

Når flere operatorer brukes i samme uttrykk, er rekkefølgen `not`, deretter `and`, og til slutt `or`. Parenteser gjør uttrykkene tydeligere.

## If-setninger og programflyt

En `if`-setning utfører en innrykket kodeblokk bare når betingelsen er sann. Med `else` kan vi velge en alternativ blokk når betingelsen er usann. Med `elif` kan vi velge mellom flere gjensidig utelukkende alternativer.

- [Virkemåte for `if`, `if-else` og `if-elif-else`](FP_21_if_basis_mekanisme.ipynb)
- [Velge type `if`-setning](FP_22_velge_if_setning.ipynb)

Husk: I en `if`-`elif`-`else`-struktur blir bare den første grenen med sann betingelse utført. Når flere handlinger skal kunne utføres samtidig, trenger vi flere uavhengige `if`-setninger.

## Et lite eksempel

Denne cellen viser en `if`-`elif`-`else`-struktur. Endre `temperatur`, og forutsi meldingen før du kjører den.

```python
temperatur = 12

if temperatur < 0:
    melding = "Frost"
elif temperatur < 15:
    melding = "Kjolig"
else:
    melding = "Varmt"

print(melding)
```

## Sammensatte og nøstede betingelser

Med `and`, `or` og `not` kan en kombinere flere tester i én betingelse. Python evaluerer fra venstre mot høyre og bruker kortslutningsevaluering: den stopper så snart resten av uttrykket ikke kan endre resultatet.

Se [sammensatte `if`-setninger](FP_23_sammensatte_if_setninger.ipynb) for intervaller, kombinasjoner av betingelser og kortslutningsevaluering.

En `if`-setning kan også ligge inni en annen `if`-setning. Da blir den indre betingelsen bare vurdert dersom den ytre grenen blir valgt. Innrykket viser hvilke linjer som hører til hvilken blokk. Se [nøstede `if`-setninger](FP_24_nøstede_if_setinger.ipynb).
