# Listegenerering: En annen måte å bruke for-løkker
En måte å bygge lister på er å bygge dem element for element med en for-løkke:

```python
# Lag liste med kvadrattall fra 1 til 100

kvadrattall = []                # Start med en tom liste
for i in range(1, 11):          # Gå gjennom tallene fra 1 til 10
    kvadrattall.append(i**2)    # Legg til kvadrattallet i listen

print(kvadrattall)              # Skriv ut listen
```


Dette er en annen måte å gjøre det på, kalt *listegenerering* eller *liste-forståelse*.

Vi bruker en for-løkke, men i stedet for å ha en kode-blokk som utføres for hver iterasjon, så beregnes et uttrykk for hver iterasjon. Uttrykket settes foran for-løkka. Disse objektene må så "samles" i en liste eller et tuppel (og mengder og ordbøker som vi ikke har hatt om ennå)


```python
kvadrattall = [i**2 for i in range(1, 11)]
```

## Filtrering av elementer i listefortåelse
Vi kan legge på en betingelse for at vi skal ha med et element i listen


```python
# Her lager vi en liste med kvadrattall for partall
kvadrattall = [i**2 for i in range(1, 11) if i % 2 == 0]

# Dette er selvfølgelig ikke nødvendig, vi kan også skrive
kvadrattall = [i**2 for i in range(2, 11, 2)]
```

Vi kan bruke denne mekanismen også for å lage tuper, men da må vi skrive tuple. Skriver vi bare paranteser rundt, får vi en *generator*, som vi kan bruke på samme måte som range().


```python
kvadrat_som_tuppel = tuple(i**2 for i in range(1, 11))
print(kvadrat_som_tuppel)


kvadrat_generator = (i**2 for i in range(1, 11))
print(kvadrat_generator)    # Vil ikke skrive ut tall, men en referanse til en generator

for i in kvadrat_generator:
    print(i, end=", ")                # Skriver ut kvadrattallene
```
