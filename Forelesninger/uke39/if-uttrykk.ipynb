## If-uttykk
Et *uttrykk* er noe som gir en verdi. Et if-uttrykk er forskjellig fra en if-setning.

La oss tenke at vi ønsker å tilordne en variabel en verdi, men hva vi ønsker å tilordne avhenger av en betingelse:

```python
n = 5
if n % 2 == 0:
    n = n // 2
else:
    n = 3*n + 1

print(n)
```

Dette kan i stedet utføres med et if-else **uttrykk**. Vi legger til en betingelse etter uttrykket, og så en else for alternativ verdi hvis betingelsen ikke er sann:


```python
n = 5
n = n // 2 if n%2 == 0 else 3*n + 1

print(n)
```
