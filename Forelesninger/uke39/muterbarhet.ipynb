# Muterbarhet

- Ikke muterbare: int, float, bool, string, tuple
- Muterbare: list, numpy.array, set, dict, m.fl.

**En variabel som referer til et ikke-muterbart objekt, kan kun endre verdi ved at variabelen tilordnes ny verdi.**

I eksemplet under vil doble_in_place og doble_ny fungere likt for ikke-multerbare datatyper.

En muterbar datatype kan endres "in place", dvs. uten at det opprettes et nytt objekt. En variabel som er tilordnet dette objektet, kan da endres uten at det skjer en tilordning til selve objektet.

Tilordning med de kombinerte tilordningsoperatorene ```+=, *=``` osv. vil ikke lage nye objekter for muterbare datatyper, mens f.eks.
```x = x * 2``` vil det.



````python
def doble_in_place(x):
    x *= 2                  # in-place, x peker fortsatt til samme objekt
    return x

def doble_ny(x):
    x = x * 2               # nytt objekt på høyresiden, som tilorndes x
    return x                # vi returnerer det NYE objektet, uten å endre input-argumentet
````

````python
liste = [1, 2, 3]               # muterbar
liste2 = liste                  # list2 peker til samme objekt som liste
print(id(liste), id(liste2))    # id() gir den unike identifikatoren for et objekt

liste2 = doble_in_place(liste2) # liste2 blir endret 
print(id(liste), id(liste2))    # og liste og liste2 peker fortsatt til samme objekt
print(liste, liste2)            # så vi får [1, 2, 3, 1, 2, 3] og [1, 2, 3, 1, 2, 3]

listea = [1, 2, 3]              # muterbar
listeb = listea             
listeb = doble_ny(listeb)       # listeb blir endret
print(listea, listeb)
print(id(listea), id(listeb))
````

## En kommenter: Return av modifisert input-argument
I ```doble_in_place``` så hadde vi strengt tatt ikke trengt å returnere x, siden vi i kallet har en referanse til input-argumentet fra der vi kaller funksjonen.

Det er noen fordeler med å likevel returnere: 
- Hvis ikke, ville funksjonen ville fungere annerledes for muterbare og ikke-muterbare input-argumenter.
- Vi gjøre det klart hva som er resultat av å kalle funksjonen.
- Input-argument trenger ikke være tilordnet en variabel før kallet.
- Vi kan kjede sammen kall.

**Det bør gjøres klart om vi endrer input-argumentet, slik at de som bruker funskjonen vet om de må lage en kopi før kallet, hvis de ønsker å bevare denne.**

Det er kanskje ikke så smart å lage en funksjon som er tiltenkt å brukes for både muterbare og ikke-muterbare data-typer?

````python
# Hvis en funksjon returnerer objektet, så kan vi gjøre endringer på det direkte:
doble_in_place(doble_in_place([1, 2, 3]))
````

````python
# Sort og sorted

liste = [3, 1, 2]
liste2 = sorted(liste) # returnerer en ny liste
print(liste2)
print(id(liste) == id(liste2))
print()

print(liste)
liste.sort()    # sorterer i listen selv, returnerer None
print(liste)    # listen er endret
````

## Flere operasjoner som lager nye objekter for muterbare datatyper



````python
liste = [1,2]
ny_liste = liste + [3, 4] # ny liste
print(ny_liste, liste)

liste += [3, 4] # endrer listen

liste.extend([5, 6]) # endrer listen, men returnerer None, så vi kan ikke tilornde det til en variabel

liste.append(7) # legger til et element i listen, returnerer også None

liste.insert(2, 0) # legger til et element på en spesifikk indeks, returnerer også None

liste.remove(0) # fjerner et element fra listen, returnerer også None

````

## NB! Slicing returnerer ny liste!

## Lister av muterbare objekter
En liste er liste av referanser til objekter. Flere referanser i listen kan referer til **samme objekt**.

````python
tekst = "abc"*5     # Uproblematisk, siden strenger er ikke-muterbare

startliste = [1,2,3]
liste = startliste*4   # Uproblematisk, kopierer ikke-muterbare objekter
print(liste)
startliste[0] = 0
print(liste)

# Liste av lister kan være problematisk, hvis vi har bruker referanser til samme liste
startliste = [1,2,3]
liste = [startliste]*2
startliste[0] = 0
print(liste) # [[0, 2, 3], [0, 2, 3]]
````
