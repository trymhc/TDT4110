## 11. Filtrering i en listeforståelse

```python
partall = [tall for tall in range(1, 11) if tall % 2 == 0]
kvadrater = [tall**2 for tall in partall]
print(kvadrater)
```

**Spørsmål:** Hvilke tall kommer med i `partall`? Når skjer beregningen av kvadratet? Hvordan kunne dette vært skrevet som én listeforståelse?

<details>
<summary>Vis mulig fasit</summary>

`if`-delen filtrerer bort oddetall, slik at `partall` blir `[2, 4, 6, 8, 10]`. Deretter kvadreres hvert av disse tallene.

Begge stegene kan skrives samlet:

```python
kvadrater = [tall**2 for tall in range(1, 11) if tall % 2 == 0]
```

</details>

## 12. Hva betyr muterbar?

Sammenlign disse to eksemplene:

```python
# Eksempel A
tall = 5
tall2 = tall
tall2 += 1
print(tall, tall2)

# Eksempel B
liste = [5]
liste2 = liste
liste2.append(1)
print(liste, liste2)
```

**Spørsmål:** Hvorfor blir resultatene forskjellige? Hvilket objekt kan endres in-place?

<details>
<summary>Vis mulig fasit</summary>

Heltall er ikke-muterbare. `tall2 += 1` lager et nytt heltallsobjekt og kobler `tall2` til det. `tall` beholder verdien 5.

Lister er muterbare. `append` endrer listeobjektet direkte, og derfor ser både `liste` og `liste2` endringen.

</details>

## 13. `+=` eller `= ... + ...`?

```python
liste = [1, 2]
liste2 = liste

liste += [3]   # samme som liste.extend([3])
print(liste)
print(liste2)
```

Hva tror du skjer hvis vi i stedet bruker:

```python
liste = liste + [3]
```

**Spørsmål:** Er `liste` og `liste2` fortsatt knyttet til samme objekt i de to tilfellene?

<details>
<summary>Vis mulig fasit</summary>

For lister endrer `liste += [3]` vanligvis listeobjektet in-place. Da ser `liste2` også endringen.

`liste = liste + [3]` lager en ny liste og tilordner den til `liste`. `liste2` peker fortsatt på den gamle listen.

Dette er en viktig forskjell mellom å endre et objekt og å koble et navn til et nytt objekt.

</details>

## 14. `sort` eller `sorted`?

```python
liste = [3, 1, 2]
ny_liste = sorted(liste)
print(liste)
print(ny_liste)

resultat = liste.sort()
print(liste)
print(resultat)
```

**Spørsmål:** Hvilken av `sorted` og `sort` lager en ny liste? Hva returnerer `sort`?

<details>
<summary>Vis mulig fasit</summary>

`sorted(liste)` lager og returnerer en ny sortert liste. Den opprinnelige listen endres ikke.

`liste.sort()` sorterer listen in-place. Metoden returnerer `None`, fordi hovedformålet er sideeffekten: å endre listen.

</details>

## 15. Hvordan lager vi en uavhengig kopi?

```python
liste = [1, 2, 3, 4]
kopi = liste[:]
kopi.append(5)
print(liste)
print(kopi)
```

**Spørsmål:** Hvorfor påvirker ikke `append` den opprinnelige listen? Hvilke andre uttrykk kan lage en grunn kopi av en liste?

<details>
<summary>Vis mulig fasit</summary>

Slicing med `liste[:]` lager en ny liste med de samme elementene. Derfor peker `liste` og `kopi` på ulike listeobjekter.

Andre vanlige måter er `liste.copy()` og `list(liste)`.

Dette er en grunn kopi, dvs. objektene er de samme.
Hvis listen inneholder muterbare objekter, må vi være oppmerksomme på at en muterende forandring av disse vil
gjenspeiles i begge listene.

</details>

## 17. Tuppel eller generator?

```python
verdier = (tall**2 for tall in range(1, 4))
print(verdier)

for verdi in verdier:
    print(verdi)
```

**Spørsmål:** Hvorfor skriver den første `print`-setningen ikke ut kvadrattallene? Hva skjer hvis vi bruker `tuple(...)` rundt uttrykket?

<details>
<summary>Vis mulig fasit</summary>

Parenteser alene lager en generator. Generatoren beregner verdiene når vi itererer over den, og `print` viser derfor bare et generatorobjekt.

`tuple(tall**2 for tall in range(1, 4))` lager tuppelen `(1, 4, 9)` med én gang.

</details>

## 18. `if`-setning eller if-uttrykk?

Disse to kodestykkene skal gi samme verdi til `resultat`:

```python
tall = 5
if tall % 2 == 0:
    resultat = tall // 2
else:
    resultat = 3 * tall + 1
```

```python
tall = 5
resultat = tall // 2 if tall % 2 == 0 else 3 * tall + 1
```

**Spørsmål:** Hva er forskjellen på en if-setning og et if-uttrykk? Når blir if-uttrykket vanskeligere å lese enn den vanlige if-setningen?

<details>
<summary>Vis mulig fasit</summary>

En if-setning styrer hvilke programsetninger som skal utføres. Et if-uttrykk beregner én verdi, og kan brukes direkte i en tilordning.

If-uttrykk passer best når valget er kort og enkelt. Hvis uttrykket blir langt, eller inneholder mye logikk, er en vanlig if-setning ofte tydeligere.

</details>

## 19. Utpakking til argumenter med `*` og `**`

En liste eller tuppel kan pakkes ut til posisjonelle argumenter med `*`:

```python
def vis_sum(a, b, c):
    print(a + b + c)

verdier = [10, 20, 30]
vis_sum(*verdier)
```

**NB! Vi har ikke hatt om ordbøker ennå**

En ordbok kan pakkes ut til navngitte argumenter med `**`:

```python
def vis_kort(navn, alder):
    print(f"{navn} er {alder} år")

person = {"navn": "Ada", "alder": 21}
vis_kort(**person)
```

**Spørsmål:** Hvilke argumenter mottar funksjonene? Hva tror du skjer hvis nøklene i `person` ikke stemmer med parameternavnene?

<details>
<summary>Vis mulig fasit</summary>

`vis_sum(*verdier)` tilsvarer:

```python
vis_sum(10, 20, 30)
```

`vis_kort(**person)` tilsvarer:

```python
vis_kort(navn="Ada", alder=21)
```

Med `*` brukes elementene i en liste eller tuppel som posisjonelle argumenter. Med `**` brukes nøklene i en ordbok som parameternavn, og verdiene blir argumentene.

Nøklene må derfor stemme med parameternavnene. Ellers får vi vanligvis en `TypeError`.

</details>

## 16. Funksjonen kan pakke argumentene med `*args`

`*` kan også brukes i funksjonsdefinisjonen for å samle flere posisjonelle argumenter i en tuppel:

```python
def vis_alle(*verdier):
    print(verdier)

vis_alle("rød", "grønn", "blå")
```

**Spørsmål:** Hva blir verdien til `verdier` inne i funksjonen? Hvordan kan vi skrive ut hvert element?

```python
for verdi in verdier:
    print(verdi)
```

Her ser vi begge retningene:

```python
verdier = [1, 2, 3]
vis_sum(*verdier)       # pakker ut ved kallet

vis_alle(1, 2, 3)       # pakker sammen i funksjonen
```

<details>
<summary>Vis mulig fasit</summary>

`verdier` blir tuppelen `("rød", "grønn", "blå")`. Vi kan iterere over den med en `for`-løkke.

Navnet `args` er vanlig, men stjernen er det som har betydning. Vi kunne også skrevet `*elementer`.

</details>
