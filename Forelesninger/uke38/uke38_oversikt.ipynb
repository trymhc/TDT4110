# Uke 38 - Funksjoner - og modularisering

Vi skal mer detaljert på hvordan funksjoner virker i Python, men ikke bare det:
Som med de fleste ting, også funksjoner, så er det en ting å lære seg hvordan de
fungerer, men en litt annen å lære seg å bruke dem!

Det er mye stoff, men vi dekker ikke alt i detalj i denne uka, men kommer til å bruke funksjoenr gjennom
resten av semesteret, og vil få alle begrepene og konseptene forklart og sett i bruk i løpet av den tiden. 

## Hvordan virker funksjoner og tilhørende begreper

- Hvordan definere funksjoner
- Hvordan kalle funksjoner
- Input: Parametre og argumenter, forskjellige måter å angi disse, standardverdier (default values)
- Returverdier vs. utskrift (`return` vs. `print`) 
- "Side-effekter" og gjenbruk av kode
- Skop: globale og lokale variabler, hvor er en variabel "synlig"
- Forskjell på Muterbare vs. ikke-muterbare objekter ved funksjonskall
- Funksjoner som objekter, og som argumenter til andre funksjoner
- Kort om docstrings og dokumentasjon, hvordan lese dokumentasjonen til en standard funksjon


## Hvorfor og hvordan bruke funksjoner

En hovedgrunn er modularisering av koden, dvs dele den opp i avgrensede deler .
Slik oppdeling skjer på forskjellige nivåer.
Dette har mange fordeler:

- Vi kan som kan lage og forstå hver del for seg
- Vi kan gjenbruke koden uten å skrive den på nytt
- Det hjelper oss å abstrahere, forstå koden på et høyere nivå
- Hjelper oss å teste: Vi kan første teste hver komponent, så teste hvordan de virker sammen

## Noen regler/prinsipper en kan bruke når man deler opp i funksjoner

- finne gjentagende kode og skille dem ut som funksjoner
- hver funksjon skal ha en vel-definert oppgave, og at koden i en funksjon "hører logisk sammen"
- hver funksjon har tilgang til kun nødvendig informasjon for å utføre sin oppgave
- lage funksjonen mer gjenbrukbar ved generalisering
