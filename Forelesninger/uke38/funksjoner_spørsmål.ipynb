## 1. Hva skjer når vi definerer og kaller en funksjon?

```python
def hilsen(navn):
    melding = f"Hei, {navn}!"
    return melding

resultat = hilsen("Ada")
print(resultat)
```

**Spørsmål:**

1. Hva er rekkefølgen som kodelinjene utføres i?
2. Hva er parameteren, og hva er argumentet?
3. Hva blir skrevet ut?
4. Hvor finnes variabelen `melding`?
5. Hvorfor returnerer funksjonen meldingen, og ikke printer den ut direkte?

<details>
<summary>Vis mulig fasit</summary>

- `def` lager funksjonen, men kjører ikke funksjonskroppen.

- `navn` er parameteren. `"Ada"` er argumentet i kallet.
- Funksjonen returnerer teksten `"Hei, Ada!"`, som deretter skrives ut.
- `melding` er lokal variabel for funksjonen og kan ikke brukes utenfor den.
- Funksjonen gjør én ting: lager en hilsen. Brukeren av funksjonen velger hva som skal skje videre. Det gjør den lettere å gjenbruke.

</details>

## 2. `print` eller `return`?

```python
def med_print(tall):
    print(tall * 2)

def med_return(tall):
    return tall * 2

a = med_print(5)
b = med_return(5)
print("a =", a)
print("b =", b)
```

**Spørsmål:** Hvorfor får `a` og `b` ulike verdier? Når er `return` mest nyttig?

<details>
<summary>Vis mulig fasit</summary>

`med_print` viser resultatet på skjermen, men returnerer ikke resultatet. Da blir `a` lik `None`.

`med_return` sender verdien tilbake til stedet der funksjonen ble kalt. Da kan vi lagre verdien, bruke den i en ny beregning eller sende den videre til en annen funksjon.

En nyttig huskeregel er: `print` er kommunikasjon med brukeren, mens `return` er kommunikasjon med resten av programmet.

</details>

## 4. Funksjoner som returnerer `True` eller `False`

```python
def er_partall(tall):
    return tall % 2 == 0

for tall in range(5):
    if er_partall(tall):
        print(tall, "er partall")
```

**Spørsmål:** Hva returnerer `er_partall(4)`? Hvorfor er en slik funksjon ofte bedre enn å skrive uttrykket direkte inne i flere `if`-setninger?

<details>
<summary>Vis mulig fasit</summary>

`er_partall(4)` returnerer `True`. Funksjonen gir et navn til kriteriet. Det gjør programmet lettere å lese, og samme sjekk kan gjenbrukes flere steder.

Dette er også et eksempel på en funksjon som utfører en beregning uten å skrive ut noe.

</details>

## 5. Flere returverdier

```python
def del_navn(fullt_navn):
    deler = fullt_navn.split()
    return deler[0], deler[-1]

fornavn, etternavn = del_navn("Ada Lovelace")
print(etternavn, fornavn)
```

**Spørsmål:** Hvor mange verdier returnerer funksjonen egentlig? Hva skjer hvis vi skriver `resultat = del_navn("Ada Lovelace")` i stedet?

<details>
<summary>Vis mulig fasit</summary>

Python returnerer én tuppel som inneholder to verdier. Ved tilordningen `fornavn, etternavn = ...` pakkes tuppelen ut i to variabler.

Med `resultat = ...` får `resultat` tuppelen `('Ada', 'Lovelace')`.

</details>

## 6. Lokale og globale variabelnavn

```python
melding = "utenfor"

def vis_melding():
    melding = "inni"
    print(melding)

vis_melding()
print(melding)
```

**Spørsmål:** Hva blir skrevet ut på de to siste linjene? Er det én eller to variabler som heter `melding`?

<details>
<summary>Vis mulig fasit</summary>

Det skrives først `inni` og deretter `utenfor`. Det er to forskjellige lokale/globalt bundne navn. Den lokale variabelen skjuler den globale mens funksjonen kjører, men endrer den ikke.

**Diskusjon:** Hvorfor kan mange globale variabler gjøre et større program vanskeligere å forstå?

</details>

## 7. Ren beregning eller sideeffekt?

Se på funksjonene under:

```python
def areal_rektangel(lengde, bredde):
    return lengde * bredde

def skriv_areal(lengde, bredde):
    print(lengde * bredde)
```

**Spørsmål:** Hvilken funksjon er enklest å teste og gjenbruke i en annen sammenheng? Hvilken funksjon har en sideeffekt?

<details>
<summary>Vis mulig fasit</summary>

`areal_rektangel` returnerer en verdi og har ingen synlig sideeffekt. Den kan brukes videre i beregninger. `skriv_areal` skriver til skjermen, og utskriften er en sideeffekt.

Begge funksjonene kan være nyttige, men de har ulike roller. Det er ofte lurt å holde beregning og presentasjon adskilt.

</details>

## 8. Hva skjer med en liste som argument?

```python
def legg_til_bonus(poeng):
    poeng.append(10)

resultater = [20, 30]
legg_til_bonus(resultater)
print(resultater)
```

**Spørsmål:** Hva blir skrevet ut? Hvorfor kan dette være annerledes enn å endre en parameter som peker på et heltall?

<details>
<summary>Vis mulig fasit</summary>

Det skrives ut `[20, 30, 10]`. Listen er et muterbart objekt, og `append` endrer selve listen som både funksjonen og kallstedet refererer til.

Et heltall er ikke muterbart. En ny tilordning av en heltallsparameter endrer derfor ikke navnet utenfor funksjonen.

</details>

## 9. Funksjoner er også objekter

```python
def doble(tall):
    return 2 * tall

beregning = doble
print(beregning(6))
```

**Spørsmål:** Hvorfor skriver vi `beregning = doble` og ikke `beregning = doble()`? Hva ville `doble()` gjort akkurat der?

**Videre:** Hva tror du denne funksjonen gjør?

```python
def bruk_beregning(beregning, tall):
    return beregning(tall)
```

<details>
<summary>Vis mulig fasit</summary>

`doble` uten parentes refererer til selve funksjonsobjektet. `doble()` forsøker å kalle funksjonen uten nødvendig argument og vil gi en feil.

`bruk_beregning` mottar en funksjon som argument og kaller den på `tall`. Dette er mulig fordi funksjoner er objekter i Python.

</details>

## 10. Hvordan leser vi dokumentasjonen?

```python
import inspect

inspect.signature(sorted)
```

**Spørsmål:** Hva tror du `/` og `*` betyr i signaturen? Hvilke argumenter kan skrives med navn?

Eksempel på signatur:

```text
sorted(iterable, /, *, key=None, reverse=False)
```

<details>
<summary>Vis mulig fasit</summary>

- `/` avslutter delen med posisjonelle-only parametre. Parametre før `/` må gis i riktig rekkefølge.
- `*` starter delen med keyword-only parametre. Parametre etter `*` må gis med navn.
- `sorted` kan derfor kalles som `sorted(data, reverse=True)`, men `reverse` må oppgis med navn.

</details>
