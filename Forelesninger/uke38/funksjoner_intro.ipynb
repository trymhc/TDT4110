## Funskjoner for å unngå duplisering av kode

- Er det noe mønster i koden under?
- Hva er likt og hva er forskjellig?

```python
# Et sted i koden
print("Week 1")
print("---------")
print("Lectures: 2")
print("Labs: 1")
print()

# et annet sted i koden
print("Week 2")
print("---------")
print("Lectures: 2")
print("Labs: 1")
print()

# et tredje sted i koden
print("Week 3")
print("---------")
print("Lectures: 2")
print("Labs: 1")
print()
```

Løkker gir en måte å unngå repetisjon, men her skal vi tenke oss at vi har lignende kode forskjellige steder i programmet, eller i forskjellige programmer. 

Vi kan pakke koden inn i en funskjon, og la det som varierer, være input til funksjonen, gjennom en *parameter*, og så kalle funksjonen

```python
def print_week_info(week_number):
    print(f"Week {week_number}")
    print("---------")
    print("Lectures: 2")
    print("Labs: 1")
    print()

print_week_info(1)
print_week_info(2)
print_week_info(3)
```

For å gjøre funksjonen mer mer generell og gjenbrukbar, f.eks. hvis det er et annet antall "lectures" eller "labs", så kan vi la disse være parametre. 

Hvis lectures vanligvis er 2, og labs 1, så kan vi gi dem standard-verdier.

```python
def print_week_info(week_number, lectures=2, labs=1):
    print(f"Week {week_number}")
    print("---------")
    print(f"Lectures: {lectures}")
    print(f"Labs: {labs}")
    print()

print_week_info(1)  # bruker standardverdier for lectures og labs
print_week_info(2, 1, 3)  # spesifiserer egne verdier for lectures og labs som posisjonelle argumenter
print_week_info(3, lectures=1, labs=3)  # spesifiserer egne verdier for lectures og labs
print_week_info(4, labs=2)  # spesifiserer kun labs, bruker standardverdi for lectures
print_week_info(5, 3)  # spesifiserer kun lectures, bruker standardverdi for labs
```
