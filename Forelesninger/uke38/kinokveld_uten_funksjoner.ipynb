# Kinokveld uten funksjoner

Dette er et litt større program før vi begynner å dele det opp i funksjoner.

Programmet skal planlegge en kinokveld for tre personer. Det skal:

- lese inn navn og alder
- bestemme billettpris ut fra alder
- spørre om personen vil ha popcorn
- beregne totalpris
- skrive ut en kvittering

Vi bruker bare begreper fra uke 34–37. Det finnes ingen egendefinerte funksjoner ennå.

```python
antall_personer = 3

print("KINOKVELD")
print("=========")

# Første person
navn1 = input("Navn på person 1: " )
alder1 = int(input("Alder på person 1: "))
if alder1 < 12:
    billett1 = 80
elif alder1 >= 67:
    billett1 = 100
else:
    billett1 = 140

popcorn1 = input("Vil " + navn1 + " ha popcorn? (ja/nei) " )
if popcorn1 == "ja":
    popcornpris1 = 45
else:
    popcornpris1 = 0

total1 = billett1 + popcornpris1

# Andre person
navn2 = input("Navn på person 2: " )
alder2 = int(input("Alder på person 2: "))
if alder2 < 12:
    billett2 = 80
elif alder2 >= 67:
    billett2 = 100
else:
    billett2 = 140

popcorn2 = input("Vil " + navn2 + " ha popcorn? (ja/nei) " )
if popcorn2 == "ja":
    popcornpris2 = 45
else:
    popcornpris2 = 0

total2 = billett2 + popcornpris2

# Tredje person
navn3 = input("Navn på person 3: " )
alder3 = int(input("Alder på person 3: "))
if alder3 < 12:
    billett3 = 80
elif alder3 >= 67:
    billett3 = 100
else:
    billett3 = 140

popcorn3 = input("Vil " + navn3 + " ha popcorn? (ja/nei) " )
if popcorn3 == "ja":
    popcornpris3 = 45
else:
    popcornpris3 = 0

total3 = billett3 + popcornpris3

# Skriv ut kvitteringen
print()
print("KVITTERING")
print("----------")
print(navn1, ":", total1, "kr")
print(navn2, ":", total2, "kr")
print(navn3, ":", total3, "kr")

sum_billetter = billett1 + billett2 + billett3
sum_popcorn = popcornpris1 + popcornpris2 + popcornpris3
sum_total = total1 + total2 + total3

print("Billetter:", sum_billetter, "kr")
print("Popcorn:", sum_popcorn, "kr")
print("Totalt:", sum_total, "kr")
```
