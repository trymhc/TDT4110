# Dekomponering i funksjoner

Dette eksemplet viser hvordan vi kan dele opp en liten oppgave i funksjoner etter prinsippene:

1. Identifiser logisk sammenhengende deler
2. Fjern duplisering og isoler enkeltoppgaver
3. Bestem hva funksjonen trenger som input og hva den skal returnere
4. Kall funksjonene riktig sted i programflyten

Målet er ikke bare å få en fungerende løsning, men å gjøre koden lettere å forstå, gjenbruke og teste.

I praksis betyr dette at vi spør oss:
- Hva er den enkelte deloppgaven?
- Hva trenger den for å gjøre jobben?
- Hva skal den gi tilbake?
- Kan vi gjøre den gjenbrukbar?

```python
# 1) Start med et enkelt problem uten funksjoner

# Vi vil lese inn to tall og skrive ut deres sum.
# Først ser vi hvordan dette kan se ut som "en stor" oppgave.

x = float(input("Skriv inn første tall: "))
y = float(input("Skriv inn andre tall: "))

sum_ = x + y
print(f"Summen er {sum_}")
```

Dette fungerer, og akkurat her ville jeg ikke si at vi har behov for å bruke funksjoner.

**MEN VI HAR STØRRE AMBISJONER!**

Vi skal gjøre det tydeligere HVA vi gjør kontra hvordan vi gjør det -> abstraksjon

Når vi f.eks. vil gjøre flere eller mere komplekse beregninger, beholder vi den generelle strukturen,og endrer kun de delene som er spesifikke for den enkelte beregning.

**Dette gjør det enklere å forstå koden, og endre og utvide koden.**

## 2) Del opp i mindre, logisk sammenhengende deler

Vi identifiserer tre oppgaver:

- lese inn data
- utføre beregning
- skrive ut resultat

Dette er et godt eksempel på abstraksjon og modularisering: Hver funksjon har én tydelig oppgave.

```python

def les_inn_tall(prompt):
    return float(input(prompt))


def beregn_sum(x, y):
    return x + y


def skriv_resultat(sum_):
    print(f"Summen er {sum_}")


x = les_inn_tall("Skriv inn første tall: ")
y = les_inn_tall("Skriv inn andre tall: ")

sum_ = beregn_sum(x, y)
skriv_resultat(sum_)
```

## 3) Hvorfor dette er bedre?

Vi kan nå:
- gjenbruke les_inn_tall() i andre program
- teste beregn_sum() isolert
- endre utskriftsformat uten å påvirke beregningen
- gjøre lesing av tall mer robust ved kun å endre les_inn_tall

Under følger et eksempel til:

```python
# 4) Et eksempel med flere steg: beregn areal av en trekant
#
# Her er selve problemstillingen delt opp i flere deloppgaver:
# - les inn grunnlinje og høyde
# - beregn areal
# - skriv ut svar


def les_inn_grunnlinje_og_hoyde():
    g = les_inn_tall("Skriv inn grunnlinjen: ")
    h = les_inn_tall("Skriv inn høyden: ")
    return g, h


def beregn_trekants_areal(grunnlinje, hoyde):
    return 0.5 * grunnlinje * hoyde


def skriv_areal(areal):
    print(f"Arealet er {areal:.2f}")


g, h = les_inn_grunnlinje_og_hoyde()
areal = beregn_trekants_areal(g, h)
skriv_areal(areal)
```

## 5) Prinsippene for oppdeling i funksjoner

1. Hver funksjon skal ha én tydelig oppgave.
2. Funksjonen skal ha tilgang til bare det den trenger.
3. Input kommer via parametre, output kommer via return.
4. Koden blir lettere å lese, teste og gjenbruke.
5. Vi skal unngå å blande beregning, innlesning og utskrift i samme funksjon.

Dette er viktig fordi funksjoner ikke bare er «kodepakker»;
de er også et verktøy for å organisere tankene bak programmet.

En god tommelfingerregel: Hvis en funksjon gjør to eller flere veldig ulike ting,
bør den kanskje deles opp i to funksjoner.

Når vi dekomponerer i funksjoner, spør vi oss:

- Hva er denne delen av programmet ansvarlig for?
- Hvilke data trenger den?
- Hva skal den returnere?
- Kan denne logikken gjenbrukes?

Hvis vi kan svare på det, er vi godt på vei til en god funksjonsdesign.

Dekomponering i funksjoner er et viktig prinsipp for å gjøre kode enklere å lese og vedlikeholde.

# Tre scaffoldede oppgaver om dekomponering i funksjoner

Her kommer tre små oppgaver som viser tre viktige prinsipper:

1. Eliminer duplisering
2. La hver funksjon ha ett ansvar
3. Øk gjenbrukbarheten

For hver oppgave er det lagt opp som en liten læringssti:
- Se på en løsning som virker, men som er litt klumpete
- Finn det som repeteres eller blandes sammen
- Del opp i funksjoner med tydelige oppgaver
- Refaktorer og vurder hvorfor koden blir bedre

Tenk alltid på disse spørsmålene:
- Hva er den enkelte deloppgaven?
- Hva trenger funksjonen for å gjøre jobben?
- Hva skal den returnere, eller skrive ut?
- Hvordan kan denne koden brukes igjen senere?


```python
# Oppgave 1: Eliminering av duplisering
#
# Mål: identifiser gjentakende kode og flytt den til en funksjon.
#
# Før: samme kode repeteres mange ganger.

print("Velkommen!")
print("Velkommen!")
print("Velkommen!")

print("Hei")
print("Hei")
print("Hei")

# Spørsmål: hva er det samme i begge delene?
# Svaret: vi skriver samme tekst flere ganger.
# Vi kan gjøre dette mer elegant med en funksjon som tar tekst som input
# og skriver den ut flere ganger.


def skriv_tre_ganger(tekst):
    for _ in range(3):
        print(tekst)


skriv_tre_ganger("Velkommen!")
skriv_tre_ganger("Hei")

# Refleksjon:
# Hvorfor er dette bedre?
# - mindre duplisering
# - enklere å endre
# - samme logikk kan brukes flere steder
```

```python
# Oppgave 2: Single responsibility
#
# Mål: få hver funksjon til å gjøre én tydelig ting.
#
# Før: innlesning, beregning og utskrift er blandet sammen.

navn = input("Skriv inn navnet ditt: ")
print("Hei", navn)
print("Lengden på navnet ditt er", len(navn))

# Her gjør samme kode tre ting:
# - les inn input
# - beregn lengde
# - skriv ut resultat
#
# Dette er et eksempel på at en funksjon eller en kodeblokk
# kan ha flere ansvar på samme tid.
# Vi deler derfor opp i mindre funksjoner med tydelig oppgave.


def les_navn():
    return input("Skriv inn navnet ditt: ")


def beregn_lengde(tekst):
    return len(tekst)


def skriv_resultat(navn, lengde):
    print(f"Hei {navn}!")
    print(f"Lengden på navnet ditt er {lengde}.")


navn = les_navn()
lengde = beregn_lengde(navn)
skriv_resultat(navn, lengde)

# Refleksjon:
# Hva er fordelene med at hver funksjon har ett ansvar?
# - enklere å forstå
# - enklere å teste
# - mindre risiko for feil i andre deler av programmet
```

```python
# Oppgave 3: Øke gjenbrukbarhet
#
# Mål: gjøre funksjonen mer generell, slik at den kan brukes i flere situasjoner.
#
# Før: funksjonen er spesifikk for én tekst.

print("Det er kaldt ute.")
print("Det er varmt ute.")
print("Det er regn ute.")

# Vi kan gjøre dette mer generisk ved å parameterisere:
# - hva teksten skal være
# - hvor mange ganger den skal skrives ut
#
# Nå blir funksjonen mer generell og gjenbrukbar.


def skriv_værmelding(tekst, antall=3):
    for _ in range(antall):
        print(tekst)


skriv_værmelding("Det er kaldt ute.")
skriv_værmelding("Det er varmt ute.", 2)
skriv_værmelding("Det er regn ute.", 4)

# Refleksjon:
# Hvordan har vi brukt parametre for å gjøre funksjonen mer generell?
# - tekst er variabel
# - antall er variabel
# - standardverdi gjør det enkelt å bruke funksjonen i mange situasjoner
```

```python
# Oppsummering: hva har vi lært?
#
# 1. Eliminering av duplisering:
#    - identifiser gjentagende kode og flytt den til en funksjon
#
# 2. Single responsibility:
#    - hver funksjon skal gjøre én tydelig oppgave
#
# 3. Økt gjenbrukbarhet:
#    - gjør funksjonen mer generell med parametre og standardverdier
#
# Når vi dekomponerer et program, er målet ikke bare å få det til å fungere,
# men også å gjøre koden enklere å lese, teste og bruke igjen.
#
# En god tommelfingerregel:
# - Hvis en funksjon gjør flere helt ulike ting, bør den kanskje deles opp.
# - Hvis samme kode oppstår flere ganger, bør den kanskje være en funksjon.
# - Hvis en funksjon kan brukes i flere sammenhenger, bør den ha parametre.

print("Dekomponering i funksjoner handler om å gjøre koden mer klar, mindre duplisert og lettere å bruke igjen.")
```
