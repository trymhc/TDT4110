### a)

Lag et program som ber brukeren om opplysninger og svarer om inntekten er skattepliktig eller skattefri. 

Regler som må implementeres:

* Hvis du bruker minst halvparten av boligen du eier til eget bruk, beregnet etter utleieverdi, er det skattefritt å leie ut resten.
* Leier du ut mer enn halvparten av egen bolig, men for under 20 000 kr i året er det også skattefritt.
* Leier du ut hele eller mer enn halvparten av egen bolig for over 20 000 kr i året er samtlige leieinntekter for hele året skattepliktige. 

**Eksempel på kjøring av kode:**
 
```python
>>>skattepliktig_beløp(19500, 60)

INFO: Denne funksjonen besvarer om utleie av boligen din er skattepliktig.
Skattepliktig beløp er: 0kr
```


```python
>>>skattepliktig_beløp(60000, 55)

INFO: Denne funksjonen besvarer om utleie av boligen din er skattepliktig.
Skattepliktig beløp er: 60000kr
```

````python
"""
* Hvis du bruker minst halvparten av boligen du eier til eget bruk, beregnet etter utleieverdi, er det skattefritt å leie ut resten.
* Leier du ut mer enn halvparten av egen bolig, men for under 20 000 kr i året er det også skattefritt.
* Leier du ut hele eller mer enn halvparten av egen bolig for over 20 000 kr i året er samtlige leieinntekter for hele året skattepliktige. 

Funksjon som tar inn årlig leieinntekt og hvor mange prosent av leiligheten som er utleid
og returnerer skyldig skattebeløp i antall kroner
"""
def skattepliktig_beløp(leieinntekt, prosent_utleid):
    print('INFO: Denne funksjonen besvarer om utleie av boligen din er skattepliktig.')
    
    beløp = int(leieinntekt)

    if prosent_utleid <= 50:
        beløp = 0
    elif leieinntekt < 20000:
        beløp = 0
    else:
        beløp = beløp
    
    return f"Skattepliktig beløp er: {beløp}kr"

skattepliktig_beløp(19500, 60)

skattepliktig_beløp(60000, 55)
````

### b)

For å leie ut *sekundærbolig* eller *fritidsbolig* gjelder det særskilte regler. Lag en tilsvarende funksjon som dekker disse behovene. (Se samme nettside for mer informasjon om ønskelig.)

Regler som må implementeres:

* Sekundærbolig:
 * Utleie av sekundærbolig beskattes fra første krone.
* Fritidsbolig:
 * Der du helt eller delvis bruker fritidsboligen til fritidsformål, og selv bruker eiendommen i rimelig omfang over tid, så vil utleieinntekter inntil kr 10 000 være skattefrie.
 * Av det overskytende beløp regnes 85 prosent som skattepliktig inntekt.
 * Dersom fritidsboligen anses som utleiehytte blir det beskatning fra første krone.
 * Om du leier ut flere enn en fritidsbolig vil grensen på 10 000 gjelde per fritidsbolig.
 
**Eksempel på kjøring av kode:**
 
```python
INFO
Dette programmet besvarer om din utleie en annen type bolig,
her sekundær- eller fritidsbolig, er skattepliktig.
Først trenger vi å vite om du leier ut en sekundær- eller en fritidsbolig.
---------------------------------------------------------------------
DATAINNHENTING:
Skriv inn type annen bolig (sekundærbolig/fritidsbolig) du har leid ut: Fritidsbolig
    
INFO
Du har valgt fritidsbolig.
Nå trenger vi først å vite om fritidsboligen(e) primært brukes til utleie eller fritid.
Deretter trenger vi å vite hvor mange fritidsbolig(er) du leier ut.
Til slutt trenger vi å vite hvor store utleieinntekter du har pr. fritidsbolig.

---------------------------------------------------------------------
DATAINNHENTING:
Skriv inn formålet med fritidsboligen(e): Fritid
Skriv inn antallet fritidsboliger du leier ut: 3
Skriv inn utleieinntekten pr. fritidsbolig: 15000
    
---------------------------------------------------------------------
SKATTEBEREGNING:
Inntekten er skattepliktig
Overskytende beløp pr. fritidsbolig er 5000
Skattepliktig inntekt pr. fritidsbolig er 4250
Totalt skattepliktig beløp er 12750
```

For å få til oppgaven vil det være nødvendig å bruke `input()`-funksjonen for å ta inn den nødvendige informasjonen om boligtype, og dersom det er en fritidsbolig må resten av informasjonen også innhentes.

***Skriv koden din her:***

````python
"""
* Sekundærbolig:
* Utleie av sekundærbolig beskattes fra første krone.
* Fritidsbolig:
* Der du helt eller delvis bruker fritidsboligen til fritidsformål, og selv bruker eiendommen i rimelig omfang over tid, så vil utleieinntekter inntil kr 10 000 være skattefrie.
* Av det overskytende beløp regnes 85 prosent som skattepliktig inntekt.
* Dersom fritidsboligen anses som utleiehytte blir det beskatning fra første krone.
* Om du leier ut flere enn en fritidsbolig vil grensen på 10 000 gjelde per fritidsbolig.
"""

def skattepliktig_beløp():
    print("INFO")
    print("Dette programmet besvarer om din utleie en annen type bolig, her sekundær- eller fritidsbolig, er skattepliktig.")
    print("Først trenger vi å vite om du leier ut en sekundær- eller en fritidsbolig.")
    print("------------------------------------------------------------------")
    boliger = ["sekundærbolig", "fritidsbolig"]
    boligtype = ""
    while boligtype not in boliger:
        boligtype = input("Skriv inn type annen bolig (sekundærbolig/fritidsbolig) du har leid ut: ")
    print("DATAINNHENTING:")
    print(f"Skriv inn type annen bolig (sekundærbolig/fritidsbolig) du har leid ut: {boligtype}")
    print("------------------------------------------------------------------")
    if (boligtype == boliger[0]):
        print("INFO")
        print(f"Du har valgt {boligtype}.")
        print(f"Deretter trenger vi å vite hvor mange {boligtype}(er) du leier ut.")
        print(f"Til slutt trenger vi å vite hvor store utleieinntekter du har pr. {boligtype}.")
        print("------------------------------------------------------------------")

        antall = int(input(f"Skriv in antall {boligtype} du leier ut: "))
        print(f"Skriv inn antallet {boligtype}er du leier ut: {antall}")

        utleie_inntekt_pr = int(input(f"Skriv inn utleieinntekten pr. {boligtype} i måneden: "))
        print(f"Skriv inn utleieinntekten pr. {boligtype}: {utleie_inntekt_pr} i måneden")

        print("------------------------------------------------------------------")
        print("SKATTEBEREGNING (Årlig):")
        print(f"Utleie av {boligtype} beskattes fra første krone.")

        skatt_pr = round(utleie_inntekt_pr*12)
        print(f"Skattepliktig inntekt pr. {boligtype} er {skatt_pr} kr")

        total = round(skatt_pr*antall)
        print(f"Totalt skattepliktig beløp er {total} kr")
    else:
        print("INFO")
        print(f"Du har valgt {boligtype}.")
        print(f"Nå trenger vi først å vite om {boligtype}en(e) primært brukes til utleie eller fritid.")
        print(f"Deretter trenger vi å vite hvor mange {boligtype}(er) du leier ut.")
        print(f"Til slutt trenger vi å vite hvor store utleieinntekter du har pr. {boligtype}.")

        bruk = ["fritid", "utleie"]
        formaal = input(f"Skriv inn formålet med {boligtype}en(e): ")
        print("------------------------------------------------------------------")
        print("DATAINNHENTING:")
        print(f"Skriv inn formålet med {boligtype}en(e): {formaal}")

        antall = int(input(f"Skriv in antall {boligtype} du leier ut: "))
        print(f"Skriv inn antallet {boligtype}er du leier ut: {antall}")

        utleie_inntekt_pr = int(input(f"Skriv inn utleieinntekten pr. {boligtype} i måneden: "))
        print(f"Skriv inn utleieinntekten pr. {boligtype}: {utleie_inntekt_pr} kr i måneden")
        
        print("------------------------------------------------------------------")
        print("SKATTEBEREGNING (Årlig):")
        
        if (formaal == bruk[1]):
            print(f"Siden {boligtype}en anses som utleiehytte blir det beskatning fra første krone.")
            print(f"Skattepliktig inntekt pr. {boligtype} er {utleie_inntekt_pr*12} kr")
            print(f"Totalt skattepliktig beløp er {utleie_inntekt_pr*12*antall} kr")
        else:
            if utleie_inntekt_pr > 10000:
                print("Inntekten er skattepliktig")
                overskytende = utleie_inntekt_pr - 10000
                print(f"Overskytende beløp pr. {boligtype} er {overskytende} kr")

                skatt_pr = round(overskytende*12*0.85)
                print(f"Skattepliktig inntekt pr. {boligtype} er {skatt_pr} kr")

                total = round(skatt_pr*antall)
                print(f"Totalt skattepliktig beløp er {total} kr")
            else:
                print("Inntekten er ikke skattepliktig")

skattepliktig_beløp()
````

````python
"""
* Sekundærbolig:
* Utleie av sekundærbolig beskattes fra første krone.
* Fritidsbolig:
* Der du helt eller delvis bruker fritidsboligen til fritidsformål, og selv bruker eiendommen i rimelig omfang over tid, så vil utleieinntekter inntil kr 10 000 være skattefrie.
* Av det overskytende beløp regnes 85 prosent som skattepliktig inntekt.
* Dersom fritidsboligen anses som utleiehytte blir det beskatning fra første krone.
* Om du leier ut flere enn en fritidsbolig vil grensen på 10 000 gjelde per fritidsbolig.
"""

SKILLELINJE = "------------------------------------------------------------------"
SKATTEFRI_GRENSE = 10000   # kr per år, per fritidsbolig
SKATTEPLIKTIG_ANDEL = 0.85


def les_valg(tekst, gyldige):
    """Spør til brukeren skriver et av de gyldige valgene."""
    svar = ""
    while svar not in gyldige:
        svar = input(f"{tekst} ({'/'.join(gyldige)}): ").strip().lower()
        if svar not in gyldige:
            print(f"Ugyldig svar. Skriv {' eller '.join(gyldige)}.")
    return svar


def les_heltall(tekst, minst=0):
    """Spør til brukeren skriver et heltall som er minst lik 'minst'."""
    while True:
        svar = input(tekst).strip().replace(" ", "")
        try:
            tall = int(svar)
        except ValueError:
            print("Ugyldig tall. Skriv et helt tall, f.eks. 5000.")
            continue
        if tall < minst:
            print(f"Tallet må være minst {minst}.")
            continue
        return tall


def skattepliktig_beløp():
    print("INFO")
    print("Dette programmet besvarer om din utleie av en annen type bolig, her sekundær- eller fritidsbolig, er skattepliktig.")
    print("Først trenger vi å vite om du leier ut en sekundær- eller en fritidsbolig.")
    print(SKILLELINJE)

    boliger = ["sekundærbolig", "fritidsbolig"]
    boligtype = les_valg("Skriv inn type annen bolig du har leid ut", boliger)

    print("DATAINNHENTING:")
    print(f"Skriv inn type annen bolig (sekundærbolig/fritidsbolig) du har leid ut: {boligtype}")
    print(SKILLELINJE)

    print("INFO")
    print(f"Du har valgt {boligtype}.")
    if boligtype == "fritidsbolig":
        print(f"Nå trenger vi først å vite om {boligtype}en(e) primært brukes til utleie eller fritid.")
    print(f"Deretter trenger vi å vite hvor mange {boligtype}(er) du leier ut.")
    print(f"Til slutt trenger vi å vite hvor store utleieinntekter du har pr. {boligtype}.")
    print(SKILLELINJE)

    print("DATAINNHENTING:")
    formaal = None
    if boligtype == "fritidsbolig":
        formaal = les_valg(f"Skriv inn formålet med {boligtype}en(e)", ["fritid", "utleie"])
        print(f"Skriv inn formålet med {boligtype}en(e): {formaal}")

    antall = les_heltall(f"Skriv inn antall {boligtype}er du leier ut: ", minst=1)
    print(f"Skriv inn antallet {boligtype}er du leier ut: {antall}")

    inntekt_mnd = les_heltall(f"Skriv inn utleieinntekten pr. {boligtype} i måneden: ")
    print(f"Skriv inn utleieinntekten pr. {boligtype}: {inntekt_mnd} kr i måneden")

    inntekt_aar = inntekt_mnd * 12

    print(SKILLELINJE)
    print("SKATTEBEREGNING (Årlig):")
    print(f"Årlig utleieinntekt pr. {boligtype} er {inntekt_aar} kr")

    if boligtype == "sekundærbolig":
        print(f"Utleie av {boligtype} beskattes fra første krone.")
        skatt_pr = inntekt_aar
    elif formaal == "utleie":
        print(f"Siden {boligtype}en anses som utleiehytte blir det beskatning fra første krone.")
        skatt_pr = inntekt_aar
    elif inntekt_aar > SKATTEFRI_GRENSE:
        print("Inntekten er skattepliktig")
        overskytende = inntekt_aar - SKATTEFRI_GRENSE
        print(f"Overskytende beløp pr. {boligtype} er {overskytende} kr")
        skatt_pr = round(overskytende * SKATTEPLIKTIG_ANDEL)
    else:
        print(f"Inntekten er ikke skattepliktig, siden den ikke overstiger {SKATTEFRI_GRENSE} kr i året.")
        skatt_pr = 0

    total = skatt_pr * antall
    print(f"Skattepliktig inntekt pr. {boligtype} er {skatt_pr} kr")
    print(f"Totalt skattepliktig beløp er {total} kr")


skattepliktig_beløp()
````
