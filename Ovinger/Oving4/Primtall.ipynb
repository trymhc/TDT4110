## a)

Skriv funksjonen `divisible(a, b)` som returnerer `True` dersom `a` er delelig med `b` og `False` ellers.

***Skriv koden i kodeblokken under***

````python
# Skriv koden din her

def divisible(a, b):
    if a % b == 0:
        return True
    else:
        return False
````

```
True
False
```

````python
# TESTKODE FOR divisible
print(divisible(10,5))
print(divisible(10,3))
````

## b)

Lag funksjonen `isPrime` som tar inn et tall `a`. Funksjonen skal ha en for-løkke som itererer fra *b = 2,3,...,a-1* og bruke funksjonen fra forrige oppgave for å sjekke om `a` er delelig med `b`. Om de er delelige avslutter du og returnerer `False`. Ellers skal du returnere `True`.

***Skriv koden i kodeblokken under***

````python
# Skriv koden din her

def isPrime(a):

    for b in range(2, a - 1, 1):
        if divisible(a, b):
            return False
    return True
        
````

Sjekk at funksjonen din fungerer med testkoden under. Output fra denne burde være:
```
True
False
```

Test gjerne også med andre verdier

````python
# TESTKODE FOR isPrime
print(isPrime(11))
print(isPrime(15))

print(isPrime(50000000000000051)) # tar mer enn 10 minutter
````

## c)

Algoritmen skal nå gjøres raskere ved hjelp av 2 små inngrep som beskrevet nedenfor. Kall den nye funksjonen for isPrime2 og ta utgangspunkt i koden du skrev i b).

- Du skal nå bare iterere deg gjennom oddetall. Men du må fortsatt sjekke om a er delelig med 2.
- Istedenfor å gå helt til $i=a−1$, skal du nå avslutte når $i > \sqrt{a}$

***Skriv koden din i kodeblokken under og sjekk at den fungerer***

````python
# Skriv koden din her

def isPrime2(a):

    if divisible(a, 2) and (a > 2):
        return False
    
    elif divisible(a, 2) and (a == 2):
        return True
    
    else:
        for b in range(3, a - 1, 2):
            if divisible(a, b):
                return False
            elif b > a**0.5:
                break
        return True

print(isPrime2(2))

print(isPrime2(11))
print(isPrime2(15))

print(isPrime2(50000000000000051)) # tar bare 16 sekunder

````
