Det er ingenting i denne øvingen som skal leveres.

Målet med denne øvingen todelt: 

- Å ha noen dokumenter til å bruke til å sjekke om "ting funker"
- Komme i gang og bli kjent med Jupyter notebooks NB! Du trenger ikke forstå all koden i notatboken!

## Om "ting funker"

Hvis du kjører dette via lenken i Canvas til Oving0, så åpnes en nettside med Jupyter, så skal alt fungere. Øving0 vil kopieres til ditt eget område på Jupyter-serveren. Der kan du jobbe med ditt egen private kopi, opprette egne filer osv. 

Hvis du vil kjøre på egen maskin, så må du installere programvare, se veiledninger på Canvas, og så teste om ting fungerer. 

** NB! Det gjør ikke noe om du ikke forstår koden ennå! **
