# Matplotlib og Numpy

I Python-verden finnes det mange biblioteker med kode man kan installere. Et bibliotek er en samling med kode som andre har skrevet, og som du kan bruke i ditt eget program. Nå skal vi sjekke om du har installert biblioteket  `numpy`, et populært for matematiske og numeriske beregninger. Kjør koden under for å se om `numpy` er installert.

```python
import numpy as np

print('Numpy er installert:', np.__version__)
print()
print('Hvis programmet ikke har krasjet, er alt i orden med numpy.')
```

`numpy` kan brukes til diverse matematiske operasjoner, som for eksempel å lage matriser og kjøre diverse matematiske operasjoner. 

```python
import numpy as np

numpy_array = np.array([1, 2, 3, 4, 5])
print('Her er et eksempel på en numpy-array:', numpy_array)

print('Du kan bruke numpy til å gjøre matematiske operasjoner på arrayen, for eksempel:')
print('Summen av elementene i arrayen:', np.sum(numpy_array))
print('Gjennomsnittet av elementene i arrayen:', np.mean(numpy_array))
print('Maksimumsverdien i arrayen:', np.max(numpy_array))
print('Minimumsverdien i arrayen:', np.min(numpy_array))
print('Standardavviket til elementene i arrayen:', np.std(numpy_array))
```

Et annet bibliotek vi skal bruke i dette kurset er `matplotlib`, som brukes til å lage grafer og visualiseringer. Kjør koden under for å se om `matplotlib` er installert.

```python
import matplotlib

print('Matplotlib er installert:', matplotlib.__version__)

print('Hvis programmet ikke har krasjet, er alt i orden med matplotlib.')
```

```python
import matplotlib.pyplot as plt
import numpy as np

print('Her er et eksempel på hvordan du kan lage en enkel linjediagram ved hjelp av matplotlib:')

x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.title('Sinuskurve')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid()
plt.show()
```
