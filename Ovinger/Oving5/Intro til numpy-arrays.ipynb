# Numpy Arrays

**Læringsmål:**

* Lister
* Numpy
* Plotting

Denne oppgaven er obligatorisk, og er en introduksjon til **numpy arrays**. Du har allerede brukt vanlige Python-lister. NumPy-arrays ligner på lister, men er spesielt laget for å regne effektivt med mange tall.

I denne oppgaven skal vi se på billettsalg for en kino. Vi bruker først vanlige lister, og deretter NumPy-arrays for å gjøre beregninger på alle tallene samtidig.

Den viktigste forskjellen vi skal se på er at regneoperasjoner på NumPy-arrays skjer **elementvis**. Det betyr at en operasjon som **array * 2** ganger hvert element i arrayen med 2.

For å bruke NumPy importerer vi biblioteket slik:

```python
import numpy as np
```

Da kan vi skrive **np.array()** i stedet for **numpy.array()**.




### a)



Kinoen har registrert hvor mange billetter som ble solgt hver dag i to uker:

```python
solgte_billetter = [74, 89, 65, 102, 118, 156, 171, 81, 95, 77, 110, 124, 164, 183]
```

Lag listen **solgte_billetter**. Lag deretter en NumPy-array som heter **solgte_billetter_np** med de samme verdiene.

Skriv ut både listen og NumPy-arrayen.

***Skriv koden din i boksen under.***



````python
import numpy as np

solgte_billetter = [74, 89, 65, 102, 118, 156, 171, 81, 95, 77, 110, 124, 164, 183]
solgte_billetter_np = np.array(solgte_billetter)
````

### c)



Kinoen regner en dag som en god salgsdag dersom det ble solgt mer enn 100 billetter.

Lag en variabel **gode_dager** som viser hvilke dager som hadde mer enn 100 solgte billetter.

Koden under gir en array med verdiene **True** og **False**. 

```python
solgte_billetter_np > 100
```

Kan du bruke dette til å hente ut alle billettsalgene som var over 100? 

***Skriv koden din i boksen under.***



````python
gode_dager = []

print("Dager som solgte mer enn 100 billetter: ")

for x in range(len(solgte_billetter_np)):
    y = solgte_billetter_np[x]
    if solgte_billetter_np[x] > 100:
        print(f"Dag {x} - {solgte_billetter_np[x]}")
        gode_dager.append(y)
gode_dager = np.array(gode_dager)

print(gode_dager)
````

<summary><b>Hint</b></summary>

```python

```

</details>

### d)



Vi kan også bruke NumPy til å regne med vektorer.

På en bestemt dag solgte kinoen tre typer billetter:

```python
antall = [82, 31, 17]
priser = [120, 90, 60]
```

Listene betyr at kinoen solgte:

* 82 voksenbilletter til 120 kroner
* 31 barnebilletter til 90 kroner
* 17 rabattbilletter til 60 kroner

Bruk funksjonene **np.array()** og **np.dot()** til å finne den totale inntekten denne dagen.

***Skriv koden din i boksen under.***



````python
antall = [82, 31, 17]
priser = [120, 90, 60]

antall_np = np.array(antall)
priser_np = np.array(priser)

total = int(np.dot(antall_np, priser_np))
print(total)

manuell = 82*120 + 31*90 + 17*60

manuell == total
````

<details>
<summary><b>Hint</b></summary>

Prikkproduktet av **antall** og **priser** regner ut

```python
82 * 120 + 31 * 90 + 17 * 60
```
</details>


### e)



Billettsalget i deloppgave a) gjelder for 14 dager, altså to uker.

Bruk **reshape()** til å lage en matrise med 2 rader og 7 kolonner fra **solgte_billetter_np**. Hver rad skal være én uke. Kall denne matrisen **salg_per_uke**.

Deretter skal du bruke **np.transpose()** til å snu matrisen, slik at hver rad viser samme ukedag i uke 1 og uke 2. Kall denne matrisen **salg_per_ukedag**.

Eksempel på formatet til resultatet:

```python
print(salg_per_uke)
>>> [[ dag 1 i uke 1, dag 2 i uke 1 ... dag 7 uke i 1],
     [ dag 1 i uke 2, dag 2 i uke 2 ... dag 7 uke i 2]]


print(salg_per_ukedag)
>>> [[ dag 1 i uke 1, dag 1 i uke 2],
     [ dag 2 i uke 1, dag 2 i uke 2],
     ...
     [ dag 7 i uke 1, dag 7 i uke 2]]
```

***Skriv koden din i boksen under.***



````python
salg_per_uke = solgte_billetter_np.reshape(2, 7)

print(salg_per_uke)

salg_per_ukedag = salg_per_uke.transpose()

print(salg_per_ukedag)
````

<details>
<summary><b>Hint</b></summary>

Hvis `arr` har 14 elementer, kan du gjøre den om til en matrise med 2 rader og 7 kolonner slik:

```python
matrise = arr.reshape(2, 7)
```

Du kan transponere en matrise slik:

```python
np.transpose(matrise)
```
</details>


### f)



Til slutt skal du plotte billettsalget for de to ukene.

Bruk matrisen **salg_per_uke** fra deloppgave e). Første rad i matrisen er uke 1, og andre rad er uke 2.

I kodecellen under får du skjelettkode for selve plottingen. Fyll inn der det står 
"**...**"" basert på kommentarene i koden.

***Skriv koden din i boksen under.***



````python
# LF
import matplotlib.pyplot as plt

# Lager dagnummer fra 1 til 7
dager = [1, 2, 3, 4, 5, 6, 7]

# Henter ut salget for uke 1 og uke 2 fra salg_per_uke
uke_1 = salg_per_uke[0]
uke_2 = salg_per_uke[1]

# Plotter billettsalget for uke 1
plt.plot(dager, uke_1, label="Uke 1")

# Plotter billettsalget for uke 2
plt.plot(dager, uke_2, label="Uke 2")

# Vannrett linje som viser grensen for en god salgsdag
plt.axhline(100, linestyle="--", color="gray", label="100 billetter")

plt.xlabel("Dag i uken")
plt.ylabel("Solgte billetter")
plt.legend()
plt.show()
````
